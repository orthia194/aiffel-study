```python
import re
from transformers import pipeline

# ====================================================================
# [1] 한글 자모 분해 및 겹받침/노이즈 정규화 함수
# ====================================================================
CHOSUNG = ['ㄱ', 'ㄲ', 'ㄴ', 'ㄷ', 'ㄸ', 'ㄹ', 'ㅁ', 'ㅂ', 'ㅃ', 'ㅅ', 'ㅆ', 'ㅇ', 'ㅈ', 'ㅉ', 'ㅊ', 'ㅋ', 'ㅌ', 'ㅍ', 'ㅎ']
JUNGSUNG = ['ㅏ', 'ㅐ', 'ㅑ', 'ㅒ', 'ㅓ', 'ㅔ', 'ㅕ', 'ㅖ', 'ㅗ', 'ㅘ', 'ㅙ', 'ㅚ', 'ㅛ', 'ㅜ', 'ㅝ', 'ㅞ', 'ㅟ', 'ㅠ', 'ㅡ', 'ㅢ', 'ㅣ']
JONGSUNG = ['', 'ㄱ', 'ㄲ', 'ㄳ', 'ㄴ', 'ㄵ', 'ㄶ', 'ㄷ', 'ㄹ', 'ㄺ', 'ㄻ', 'ㄼ', 'ㄽ', 'ㄾ', 'ㄿ', 'ㅀ', 'ㅁ', 'ㅂ', 'ㅄ', 'ㅅ', 'ㅆ', 'ㅇ', 'ㅈ', 'ㅊ', 'ㅋ', 'ㅌ', 'ㅍ', 'ㅎ']

# 우회용으로 자주 쓰이는 겹받침 단순화 매핑
COMPLEX_BATCHIM_MAP = {
    'ㄼ': '',    # 씗 -> 씨
    'ㅀ': '',    # 싷 -> 시
    'ㄶ': '',    # 싾 -> 싸
    'ㄻ': 'ㄹ',  # 빪 -> 빨, 밞 -> 발
    'ㄺ': 'ㄱ',
    'ㄳ': 'ㄱ',
    'ㅄ': 'ㅂ'
}

def decompose_hangul(char):
    """한글 한 글자를 초성, 중성, 종성으로 분해"""
    if '가' <= char <= '힣':
        code = ord(char) - 0xAC00
        cho = code // (21 * 28)
        jung = (code % (21 * 28)) // 28
        jong = code % 28
        return CHOSUNG[cho], JUNGSUNG[jung], JONGSUNG[jong]
    return char, '', ''

def compose_hangul(cho, jung, jong=''):
    """초성, 중성, 종성을 하나의 한글 글자로 결합"""
    if cho in CHOSUNG and jung in JUNGSUNG:
        cho_idx = CHOSUNG.index(cho)
        jung_idx = JUNGSUNG.index(jung)
        jong_idx = JONGSUNG.index(jong) if jong in JONGSUNG else 0
        return chr(0xAC00 + (cho_idx * 21 + jung_idx) * 28 + jong_idx)
    return cho

def clean_bypassed_text(text):
    """변형 욕설(특수문자, 이상한 겹받침)을 원형으로 복원하는 전처리 엔진"""
    # 1. 글자 사이에 끼워넣은 숫자나 특수문자 제거 (예: 시1발 -> 시발)
    text = re.sub(r'[\d~!@#$%^&*()_+|<>?:{}]', '', text)
    
    # 2. 겹받침 노이즈 정규화 (예: 씗 -> 씨, 빪 -> 빨)
    normalized_chars = []
    for char in text:
        if '가' <= char <= '힣':
            cho, jung, jong = decompose_hangul(char)
            # 겹받침이 들어가 있다면 단순화
            if jong in COMPLEX_BATCHIM_MAP:
                jong = COMPLEX_BATCHIM_MAP[jong]
            normalized_chars.append(compose_hangul(cho, jung, jong))
        else:
            normalized_chars.append(char)
            
    cleaned_text = "".join(normalized_chars)
    
    # 3. 흔한 쌍자음 변형 원형 복원 (씨빨 -> 씨발)
    cleaned_text = cleaned_text.replace("씨빨", "씨발").replace("시빨", "시발")
    return cleaned_text


# ====================================================================
# [2] 모델 로딩 및 전처리 전/후 비교 테스트
# ====================================================================
print("⏳ 모델 로딩 중...")
classifier = pipeline("text-classification", model="smilegate-ai/kor_unsmile", top_k=None)
print("✅ 로딩 완료!\n")

def check_text(text):
    results = classifier(text)
    for res in results[0]:
        if res['label'] == '악플/욕설':
            return res['score'] * 100
    return 0.0

# ⭐️ 아까 감지 실패했던 '씗빪' 테스트 문장
test_sentence = "씳빪아"

print("=" * 60)
print(f"📌 [테스트 입력 문장]: '{test_sentence}'")
print("=" * 60)

# 1. 전처리 없이 그냥 넣었을 때
score_before = check_text(test_sentence)
print(f"❌ 1. 전처리 전 (기존 모델만 사용):")
print(f"   - 감지된 악플/욕설 확률: {score_before:.2f}%")
if score_before < 50:
    print("   👉 [결과: 뚫림 (감지 실패!)]")

print("-" * 60)

# 2. 우리 전처리 엔진을 거쳐서 넣었을 때
preprocessed_sentence = clean_bypassed_text(test_sentence)
score_after = check_text(preprocessed_sentence)
print(f"⭕ 2. 우리 전처리 엔진 통과 후:")
print(f"   - 복원된 문장: '{preprocessed_sentence}'")
print(f"   - 감지된 악플/욕설 확률: {score_after:.2f}%")
if score_after >= 50:
    print("   👉 🚨 [결과: 완벽 감지 및 검거 성공!]")
print("=" * 60)
```

```python
import re
from transformers import pipeline

# ====================================================================
# [1] 자모 결합 및 겹받침 정규화 엔진
# ====================================================================
CHOSUNG = ['ㄱ', 'ㄲ', 'ㄴ', 'ㄷ', 'ㄸ', 'ㄹ', 'ㅁ', 'ㅂ', 'ㅃ', 'ㅅ', 'ㅆ', 'ㅇ', 'ㅈ', 'ㅉ', 'ㅊ', 'ㅋ', 'ㅌ', 'ㅍ', 'ㅎ']
JUNGSUNG = ['ㅏ', 'ㅐ', 'ㅑ', 'ㅒ', 'ㅓ', 'ㅔ', 'ㅕ', 'ㅖ', 'ㅗ', 'ㅘ', 'ㅙ', 'ㅚ', 'ㅛ', 'ㅜ', 'ㅝ', 'ㅞ', 'ㅟ', 'ㅠ', 'ㅡ', 'ㅢ', 'ㅣ']
JONGSUNG = ['', 'ㄱ', 'ㄲ', 'ㄳ', 'ㄴ', 'ㄵ', 'ㄶ', 'ㄷ', 'ㄹ', 'ㄺ', 'ㄻ', 'ㄼ', 'ㄽ', 'ㄾ', 'ㄿ', 'ㅀ', 'ㅁ', 'ㅂ', 'ㅄ', 'ㅅ', 'ㅆ', 'ㅇ', 'ㅈ', 'ㅊ', 'ㅋ', 'ㅌ', 'ㅍ', 'ㅎ']

COMPLEX_BATCHIM_MAP = {
    'ㄼ': '',    # 씗 -> 씨
    'ㅀ': '',    # 싷 -> 시
    'ㄶ': '',    # 싾 -> 싸
    'ㄻ': 'ㄹ',  # 빪 -> 빨, 밞 -> 발
    'ㄺ': 'ㄱ',
    'ㄳ': 'ㄱ',
    'ㅄ': 'ㅂ'
}

def compose_syllable(cho, jung, jong=''):
    """초성, 중성, 종성을 하나의 완성된 한글 글자로 결합"""
    c_idx = CHOSUNG.index(cho)
    j_idx = JUNGSUNG.index(jung)
    t_idx = JONGSUNG.index(jong) if jong in JONGSUNG else 0
    return chr(0xAC00 + (c_idx * 21 + j_idx) * 28 + t_idx)

def combine_jamo(text):
    """쪼개진 자모(ㅅ, ㅣ, ㅂ, ㅏ, ㄹ)를 완성형 글자(시, 발)로 조립하는 오토마타"""
    chars = list(text)
    result = []
    i = 0
    n = len(chars)
    
    while i < n:
        c = chars[i]
        # 초성 + 중성이 연속으로 등장하는지 확인
        if c in CHOSUNG and i + 1 < n and chars[i+1] in JUNGSUNG:
            cho = c
            jung = chars[i+1]
            jong = ""
            
            # 다음 글자가 종성(받침) 후보인지 확인
            if i + 2 < n and chars[i+2] in JONGSUNG and chars[i+2] != "":
                # 만약 그 다다음 글자가 또 모음(중성)이라면, 이 받침은 다음 글자의 초성으로 양보!
                # 예: 'ㅅ','ㅣ','ㅂ','ㅏ','ㄹ' 에서 'ㅂ'은 '시'의 받침이 아니라 '발'의 초성임
                if i + 3 < n and chars[i+3] in JUNGSUNG:
                    jong = ""
                    i += 2
                else:
                    jong = chars[i+2]
                    i += 3
            else:
                i += 2
                
            result.append(compose_syllable(cho, jung, jong))
        else:
            result.append(c)
            i += 1
            
    return "".join(result)

def clean_bypassed_text(text):
    """특수문자 제거 + 자모 결합 + 겹받침 정규화를 한 번에 수행"""
    # 1. 자모 사이에 끼어든 숫자/특수문자 제거 (예: ㅅㅣ1발 -> ㅅㅣ발)
    text = re.sub(r'[\d~!@#$%^&*()_+|<>?:{}\[\]\-]', '', text)
    
    # 2. 쪼개진 자모를 완성형 글자로 결합 (예: ㅅㅣ발 -> 시발, ㅈㅗㅈ -> 좆)
    text = combine_jamo(text)
    
    # 3. 씗빪 같은 이상한 겹받침 분해 후 단순화
    normalized_chars = []
    for char in text:
        if '가' <= char <= '힣':
            code = ord(char) - 0xAC00
            cho = CHOSUNG[code // (21 * 28)]
            jung = JUNGSUNG[(code % (21 * 28)) // 28]
            jong = JONGSUNG[code % 28]
            
            if jong in COMPLEX_BATCHIM_MAP:
                jong = COMPLEX_BATCHIM_MAP[jong]
            normalized_chars.append(compose_syllable(cho, jung, jong))
        else:
            normalized_chars.append(char)
            
    cleaned_text = "".join(normalized_chars)
    
    # 4. 발음 변형 마무리
    cleaned_text = cleaned_text.replace("씨빨", "씨발").replace("시빨", "시발")
    return cleaned_text


# ====================================================================
# [2] 아까 뚫렸던 문장들 전처리 전/후 비교 테스트
# ====================================================================
print("⏳ 모델 로딩 중...")
classifier = pipeline("text-classification", model="smilegate-ai/kor_unsmile", top_k=None)
print("✅ 로딩 완료!\n")

def get_swear_score(text):
    results = classifier(text)
    for res in results[0]:
        if res['label'] == '악플/욕설':
            return res['score'] * 100
    return 0.0

# 🧪 아까 뚫리거나 위태로웠던 문장 4인방
test_sentences = [
    "야 이 ㅅㅣ1발롬아",
    "ㅅㅣㅂㅏㄹ 저리 가라",
    "ㅈㅗㅈ까고 있네 진짜",
    "야 이 씗빪아",
    "시...발...새...끼",
    "조 까 구 있 네",
    "개@새@끼",
    "시ㅂㅏㄹ련아",
    "ㅅ1ㅂㅏㄹ",
    "꺼 ㅈㅕ",
    "시베리아 귤껍데기",
    "개색구 야",
    "신발끈 조카신발",
    "수박 씨 발라먹어",
    "tlqkf",
    "shdem",
    "개ㅅH끼",
    "앰창",
    "느금마",
    
]

print("=" * 80)
print(f"{'원래 문장':<20} ➔ {'복원된 문장':<18} | {'전처리 전':<10} ➔ {'전처리 후'}")
print("=" * 80)

for original in test_sentences:
    # 1. 전처리 전 점수
    score_before = get_swear_score(original)
    
    # 2. 자모 결합 엔진 통과
    restored = clean_bypassed_text(original)
    
    # 3. 전처리 후 점수
    score_after = get_swear_score(restored)
    
    status = "🚨 검거 성공!" if score_after >= 50 else "❌ 여전히 실패"
    print(f"{original:<20} ➔ {restored:<18} | {score_before:>8.2f}% ➔ {score_after:>8.2f}% | {status}")

print("=" * 80)
```

```python
import re
from transformers import pipeline

# ====================================================================
# [1] 통합 전처리 및 정규화 엔진
# ====================================================================
ENG_KEY_MAP = {
    'q': 'ㅂ', 'w': 'ㅈ', 'e': 'ㄷ', 'r': 'ㄱ', 't': 'ㅅ',
    'y': 'ㅛ', 'u': 'ㅕ', 'i': 'ㅑ', 'o': 'ㅐ', 'p': 'ㅔ',
    'a': 'ㅁ', 's': 'ㄴ', 'd': 'ㅇ', 'f': 'ㄹ', 'g': 'ㅎ',
    'h': 'ㅗ', 'j': 'ㅓ', 'k': 'ㅏ', 'l': 'ㅣ',
    'z': 'ㅋ', 'x': 'ㅌ', 'c': 'ㅊ', 'v': 'ㅍ', 'b': 'ㅠ',
    'n': 'ㅜ', 'm': 'ㅡ',
    'Q': 'ㅃ', 'W': 'ㅉ', 'E': 'ㄸ', 'R': 'ㄲ', 'T': 'ㅆ',
    'O': 'ㅒ', 'P': 'ㅖ'
}

CHOSUNG = ['ㄱ', 'ㄲ', 'ㄴ', 'ㄷ', 'ㄸ', 'ㄹ', 'ㅁ', 'ㅂ', 'ㅃ', 'ㅅ', 'ㅆ', 'ㅇ', 'ㅈ', 'ㅉ', 'ㅊ', 'ㅋ', 'ㅌ', 'ㅍ', 'ㅎ']
JUNGSUNG = ['ㅏ', 'ㅐ', 'ㅑ', 'ㅒ', 'ㅓ', 'ㅔ', 'ㅕ', 'ㅖ', 'ㅗ', 'ㅘ', 'ㅙ', 'ㅚ', 'ㅛ', 'ㅜ', 'ㅝ', 'ㅞ', 'ㅟ', 'ㅠ', 'ㅡ', 'ㅢ', 'ㅣ']
JONGSUNG = ['', 'ㄱ', 'ㄲ', 'ㄳ', 'ㄴ', 'ㄵ', 'ㄶ', 'ㄷ', 'ㄹ', 'ㄺ', 'ㄻ', 'ㄼ', 'ㄽ', 'ㄾ', 'ㄿ', 'ㅀ', 'ㅁ', 'ㅂ', 'ㅄ', 'ㅅ', 'ㅆ', 'ㅇ', 'ㅈ', 'ㅊ', 'ㅋ', 'ㅌ', 'ㅍ', 'ㅎ']

COMPLEX_BATCHIM_MAP = {
    'ㄼ': '', 'ㅀ': '', 'ㄶ': '', 'ㄻ': 'ㄹ', 'ㄺ': 'ㄱ', 'ㄳ': 'ㄱ', 'ㅄ': 'ㅂ'
}

SLANG_MAP = {
    "개색구": "개새끼", "개색기": "개새끼", "개새꾸": "개새끼",
    "시베리아 귤껍데기": "씨발"
}

# ⭐️ 명백한 패드립 직권 차단 키워드 (오탐 방지)
HARD_SWEAR_WORDS = ["느금마", "느검마", "니애미", "니애비", "너네애미"]

# ⭐️ 억울한 오탐을 막기 위한 정상 표현 보호 목록
SAFE_PHRASES = ["수박 씨", "수박씨", "조카 신발", "조카신발", "시발점", "신발끈", "애미야", "어미야"]

def convert_eng_to_jamo(text):
    words = text.split()
    converted_words = []
    for word in words:
        if re.fullmatch(r'[a-zA-Z]+', word):
            converted = "".join([ENG_KEY_MAP.get(ch, ch) for ch in word])
            converted_words.append(converted)
        else:
            converted_words.append(word)
    return " ".join(converted_words)

def compose_syllable(cho, jung, jong=''):
    c_idx = CHOSUNG.index(cho)
    j_idx = JUNGSUNG.index(jung)
    t_idx = JONGSUNG.index(jong) if jong in JONGSUNG else 0
    return chr(0xAC00 + (c_idx * 21 + j_idx) * 28 + t_idx)

def combine_jamo(text):
    chars = list(text)
    result = []
    i = 0
    n = len(chars)
    while i < n:
        c = chars[i]
        if c in CHOSUNG and i + 1 < n and chars[i+1] in JUNGSUNG:
            cho = c
            jung = chars[i+1]
            jong = ""
            if i + 2 < n and chars[i+2] in JONGSUNG and chars[i+2] != "":
                if i + 3 < n and chars[i+3] in JUNGSUNG:
                    jong = ""
                    i += 2
                else:
                    jong = chars[i+2]
                    i += 3
            else:
                i += 2
            result.append(compose_syllable(cho, jung, jong))
        else:
            result.append(c)
            i += 1
    return "".join(result)

def clean_bypassed_text(text):
    text = convert_eng_to_jamo(text)
    text = re.sub(r'[\d~!@#$%^&*()_+|<>?:{}\[\]\-.]', '', text)
    text = combine_jamo(text)
    
    normalized_chars = []
    for char in text:
        if '가' <= char <= '힣':
            code = ord(char) - 0xAC00
            cho = CHOSUNG[code // (21 * 28)]
            jung = JUNGSUNG[(code % (21 * 28)) // 28]
            jong = JONGSUNG[code % 28]
            if jong in COMPLEX_BATCHIM_MAP:
                jong = COMPLEX_BATCHIM_MAP[jong]
            normalized_chars.append(compose_syllable(cho, jung, jong))
        else:
            normalized_chars.append(char)
    text = "".join(normalized_chars)
    
    for slang, clean in SLANG_MAP.items():
        text = text.replace(slang, clean)
        
    text = text.replace("씨빨", "씨발").replace("시빨", "시발")
    return text


# ====================================================================
# [2] 모델 로딩
# ====================================================================
print("⏳ 모델 로딩 중...")
classifier = pipeline("text-classification", model="smilegate-ai/kor_unsmile", top_k=None)
print("✅ 로딩 완료!\n")

def evaluate_toxicity(text):
    results = classifier(text)[0]
    toxic_labels = [r for r in results if r['label'] != 'clean']
    top_toxic = max(toxic_labels, key=lambda x: x['score'])
    return top_toxic['label'], top_toxic['score'] * 100


# ====================================================================
# [3] 일상 대화 10개 & 대화 속 욕설 10개 데이터셋
# ====================================================================
dataset = [
    # ---------------- [A] 일상 대화 10개 (Target: 정상 통과) ----------------
    {"text": "오늘 날씨 진짜 좋다. 점심 뭐 먹으러 갈래?", "is_swear": False},
    {"text": "이번 주말에 친구들이랑 오랜만에 축구하기로 했어.", "is_swear": False},
    {"text": "프로젝트 마감 기한이 얼마 안 남아서 오늘 야근할 것 같아.", "is_swear": False},
    {"text": "새로 산 신발 신고 출근했더니 기분이 너무 상쾌하다.", "is_swear": False}, # '신발' 함정
    {"text": "할머니께서 애미야 밥 차려라 하고 부르시더라.", "is_swear": False},       # '애미야' 함정
    {"text": "수박 먹을 때 수박 씨 골라내는 게 제일 귀찮아.", "is_swear": False},     # '수박 씨' 함정
    {"text": "그 실패가 우리 팀의 새로운 시발점이 되었어.", "is_swear": False},        # '시발점' 함정
    {"text": "어제 본 영화 생각보다 너무 개재밌고 감동적이었어.", "is_swear": False},  # '개~' 긍정 칭찬
    {"text": "과제 다 끝내고 나니까 진짜 살 것 같네.", "is_swear": False},
    {"text": "내일 아침 9시까지 3층 회의실로 모여주세요.", "is_swear": False},

    # ---------------- [B] 대화 속 우회/변형 욕설 10개 (Target: 검거 성공) ----------------
    {"text": "게임하는데 뒤에서 자꾸 ㅈㅗㅈ까고 있네 진짜 짜증나게", "is_swear": True},
    {"text": "저 팀원 아무것도 안 하고 노는데 진짜 ㅂㅕㅇㅅㅣㄴ 같음", "is_swear": True},
    {"text": "출근길에 지하철 연착돼서 야 이 ㅅㅣ1발 지각하겠다", "is_swear": True},
    {"text": "팀장님이 또 말도 안 되는 일 시키는데 tlqkf 때려칠까", "is_swear": True},
    {"text": "어제 술 취해서 씗빪 거리던 놈 누구냐 진상이다", "is_swear": True},
    {"text": "주차를 이따위로 해놓으면 어떡하냐 개@새@끼야", "is_swear": True},
    {"text": "선 넘지 마라 진짜 시ㅂㅏㄹ련아 적당히 해", "is_swear": True},
    {"text": "걔 싸가지 없어서 선배들한테 욕먹고 꺼 져 소리 들음", "is_swear": True},
    {"text": "장난치는 것도 아니고 패드립으로 느금마 거리는 건 선 넘었지", "is_swear": True},
    {"text": "음식 배달 2시간 걸린 거 실화냐 사장 새끼 장난치나", "is_swear": True}
]


# ====================================================================
# [4] 종합 테스트 실행 및 성적표 산출
# ====================================================================
print("=" * 95)
print(f"{'#':<3} | {'테스트 문장':<38} | {'최고 혐오 라벨':<10} | {'확률':<7} | {'판정'}")
print("=" * 95)

clean_pass_count = 0     # 정상 대화인데 정상 통과한 개수 (성공)
swear_caught_count = 0   # 욕설인데 검거 성공한 개수 (성공)

for i, data in enumerate(dataset, 1):
    original = data["text"]
    is_swear = data["is_swear"]
    
    # 1. 패드립 직권 차단 체크
    is_hard_swear = any(bad in original for bad in HARD_SWEAR_WORDS)
    
    # 2. 화이트리스트 체크
    is_safe = any(phrase in original for phrase in SAFE_PHRASES)
    
    # 3. 전처리 파이프라인
    restored = clean_bypassed_text(original)
    
    # 4. 모델 판정
    label, score = evaluate_toxicity(restored)
    
    # 5. 최종 판정 로직
    if is_hard_swear:
        verdict_toxic = True
        label = "패드립/가족"
        score = 99.99
    elif is_safe and score < 90:
        verdict_toxic = False
    elif score >= 50:
        verdict_toxic = True
    else:
        verdict_toxic = False
        
    # 결과 평가
    if not is_swear:
        # 정상 대화 검증
        if not verdict_toxic:
            status = "✅ [정상 통과]"
            clean_pass_count += 1
        else:
            status = "⚠️ [오탐 발생 (오류)]"
    else:
        # 욕설 대화 검증
        if verdict_toxic:
            status = f"🚨 [검거 성공 ({label})]"
            swear_caught_count += 1
        else:
            status = "❌ [감지 실패 (뚫림)]"

    # 너무 긴 문장은 화면을 위해 자름
    display_text = (original[:35] + '..') if len(original) > 35 else original
    print(f"{i:<3} | {display_text:<38} | {label:<10} | {score:>6.2f}% | {status}")

print("=" * 95)
print("📊 [최종 종합 성적표]")
print(f"1. 정상 일상 대화 보존율 (10개 중) : {clean_pass_count}/10 ({clean_pass_count*10}%)")
print(f"2. 대화 속 우회 욕설 탐지율 (10개 중) : {swear_caught_count}/10 ({swear_caught_count*10}%)")
print(f"3. 전체 종합 정확도 (20개 중)       : {clean_pass_count + swear_caught_count}/20 ({(clean_pass_count + swear_caught_count)*5}%)")
print("=" * 95)
```

```python
import re
import time  # ⭐️ 시간 측정을 위한 라이브러리 추가
from transformers import pipeline

# ====================================================================
# [1] 통합 전처리 및 정규화 엔진
# ====================================================================
ENG_KEY_MAP = {
    'q': 'ㅂ', 'w': 'ㅈ', 'e': 'ㄷ', 'r': 'ㄱ', 't': 'ㅅ',
    'y': 'ㅛ', 'u': 'ㅕ', 'i': 'ㅑ', 'o': 'ㅐ', 'p': 'ㅔ',
    'a': 'ㅁ', 's': 'ㄴ', 'd': 'ㅇ', 'f': 'ㄹ', 'g': 'ㅎ',
    'h': 'ㅗ', 'j': 'ㅓ', 'k': 'ㅏ', 'l': 'ㅣ',
    'z': 'ㅋ', 'x': 'ㅌ', 'c': 'ㅊ', 'v': 'ㅍ', 'b': 'ㅠ',
    'n': 'ㅜ', 'm': 'ㅡ',
    'Q': 'ㅃ', 'W': 'ㅉ', 'E': 'ㄸ', 'R': 'ㄲ', 'T': 'ㅆ',
    'O': 'ㅒ', 'P': 'ㅖ'
}

CHOSUNG = ['ㄱ', 'ㄲ', 'ㄴ', 'ㄷ', 'ㄸ', 'ㄹ', 'ㅁ', 'ㅂ', 'ㅃ', 'ㅅ', 'ㅆ', 'ㅇ', 'ㅈ', 'ㅉ', 'ㅊ', 'ㅋ', 'ㅌ', 'ㅍ', 'ㅎ']
JUNGSUNG = ['ㅏ', 'ㅐ', 'ㅑ', 'ㅒ', 'ㅓ', 'ㅔ', 'ㅕ', 'ㅖ', 'ㅗ', 'ㅘ', 'ㅙ', 'ㅚ', 'ㅛ', 'ㅜ', 'ㅝ', 'ㅞ', 'ㅟ', 'ㅠ', 'ㅡ', 'ㅢ', 'ㅣ']
JONGSUNG = ['', 'ㄱ', 'ㄲ', 'ㄳ', 'ㄴ', 'ㄵ', 'ㄶ', 'ㄷ', 'ㄹ', 'ㄺ', 'ㄻ', 'ㄼ', 'ㄽ', 'ㄾ', 'ㄿ', 'ㅀ', 'ㅁ', 'ㅂ', 'ㅄ', 'ㅅ', 'ㅆ', 'ㅇ', 'ㅈ', 'ㅊ', 'ㅋ', 'ㅌ', 'ㅍ', 'ㅎ']

COMPLEX_BATCHIM_MAP = {
    'ㄼ': '', 'ㅀ': '', 'ㄶ': '', 'ㄻ': 'ㄹ', 'ㄺ': 'ㄱ', 'ㄳ': 'ㄱ', 'ㅄ': 'ㅂ'
}

SLANG_MAP = {
    "개색구": "개새끼", "개색기": "개새끼", "개새꾸": "개새끼",
    "시베리아 귤껍데기": "씨발"
}

HARD_SWEAR_WORDS = ["느금마", "느검마", "니애미", "니애비", "너네애미"]
SAFE_PHRASES = ["수박 씨", "수박씨", "조카 신발", "조카신발", "시발점", "신발끈", "애미야", "어미야"]

def convert_eng_to_jamo(text):
    words = text.split()
    converted_words = []
    for word in words:
        if re.fullmatch(r'[a-zA-Z]+', word):
            converted = "".join([ENG_KEY_MAP.get(ch, ch) for ch in word])
            converted_words.append(converted)
        else:
            converted_words.append(word)
    return " ".join(converted_words)

def compose_syllable(cho, jung, jong=''):
    c_idx = CHOSUNG.index(cho)
    j_idx = JUNGSUNG.index(jung)
    t_idx = JONGSUNG.index(jong) if jong in JONGSUNG else 0
    return chr(0xAC00 + (c_idx * 21 + j_idx) * 28 + t_idx)

def combine_jamo(text):
    chars = list(text)
    result = []
    i = 0
    n = len(chars)
    while i < n:
        c = chars[i]
        if c in CHOSUNG and i + 1 < n and chars[i+1] in JUNGSUNG:
            cho = c
            jung = chars[i+1]
            jong = ""
            if i + 2 < n and chars[i+2] in JONGSUNG and chars[i+2] != "":
                if i + 3 < n and chars[i+3] in JUNGSUNG:
                    jong = ""
                    i += 2
                else:
                    jong = chars[i+2]
                    i += 3
            else:
                i += 2
            result.append(compose_syllable(cho, jung, jong))
        else:
            result.append(c)
            i += 1
    return "".join(result)

def clean_bypassed_text(text):
    text = convert_eng_to_jamo(text)
    text = re.sub(r'[\d~!@#$%^&*()_+|<>?:{}\[\]\-.]', '', text)
    text = combine_jamo(text)
    
    normalized_chars = []
    for char in text:
        if '가' <= char <= '힣':
            code = ord(char) - 0xAC00
            cho = CHOSUNG[code // (21 * 28)]
            jung = JUNGSUNG[(code % (21 * 28)) // 28]
            jong = JONGSUNG[code % 28]
            if jong in COMPLEX_BATCHIM_MAP:
                jong = COMPLEX_BATCHIM_MAP[jong]
            normalized_chars.append(compose_syllable(cho, jung, jong))
        else:
            normalized_chars.append(char)
    text = "".join(normalized_chars)
    
    for slang, clean in SLANG_MAP.items():
        text = text.replace(slang, clean)
        
    text = text.replace("씨빨", "씨발").replace("시빨", "시발")
    return text


# ====================================================================
# [2] 모델 로딩 (첫 웜업 포함)
# ====================================================================
print("⏳ 모델 로딩 중...")
classifier = pipeline("text-classification", model="smilegate-ai/kor_unsmile", top_k=None)
# 초기 워밍업(첫 실행 시 지연 방지)
_ = classifier("테스트")
print("✅ 로딩 완료!\n")

def evaluate_toxicity(text):
    results = classifier(text)[0]
    toxic_labels = [r for r in results if r['label'] != 'clean']
    top_toxic = max(toxic_labels, key=lambda x: x['score'])
    return top_toxic['label'], top_toxic['score'] * 100


# ====================================================================
# [3] 일상 대화 10개 & 대화 속 욕설 10개 데이터셋
# ====================================================================
dataset = [
    {"text": "오늘 날씨 진짜 좋다. 점심 뭐 먹으러 갈래?", "is_swear": False},
    {"text": "이번 주말에 친구들이랑 오랜만에 축구하기로 했어.", "is_swear": False},
    {"text": "프로젝트 마감 기한이 얼마 안 남아서 오늘 야근할 것 같아.", "is_swear": False},
    {"text": "새로 산 신발 신고 출근했더니 기분이 너무 상쾌하다.", "is_swear": False},
    {"text": "할머니께서 애미야 밥 차려라 하고 부르시더라.", "is_swear": False},
    {"text": "수박 먹을 때 수박 씨 골라내는 게 제일 귀찮아.", "is_swear": False},
    {"text": "그 실패가 우리 팀의 새로운 시발점이 되었어.", "is_swear": False},
    {"text": "어제 본 영화 생각보다 너무 개재밌고 감동적이었어.", "is_swear": False},
    {"text": "과제 다 끝내고 나니까 진짜 살 것 같네.", "is_swear": False},
    {"text": "내일 아침 9시까지 3층 회의실로 모여주세요.", "is_swear": False},

    {"text": "게임하는데 뒤에서 자꾸 ㅈㅗㅈ까고 있네 진짜 짜증나게", "is_swear": True},
    {"text": "저 팀원 아무것도 안 하고 노는데 진짜 ㅂㅕㅇㅅㅣㄴ 같음", "is_swear": True},
    {"text": "출근길에 지하철 연착돼서 야 이 ㅅㅣ1발 지각하겠다", "is_swear": True},
    {"text": "팀장님이 또 말도 안 되는 일 시키는데 tlqkf 때려칠까", "is_swear": True},
    {"text": "어제 술 취해서 씗빪 거리던 놈 누구냐 진상이다", "is_swear": True},
    {"text": "주차를 이따위로 해놓으면 어떡하냐 개@새@끼야", "is_swear": True},
    {"text": "선 넘지 마라 진짜 시ㅂㅏㄹ련아 적당히 해", "is_swear": True},
    {"text": "걔 싸가지 없어서 선배들한테 욕먹고 꺼 져 소리 들음", "is_swear": True},
    {"text": "장난치는 것도 아니고 패드립으로 느금마 거리는 건 선 넘었지", "is_swear": True},
    {"text": "음식 배달 2시간 걸린 거 실화냐 사장 새끼 장난치나", "is_swear": True}
]


# ====================================================================
# [4] 속도 측정 실행 및 종합 성적표
# ====================================================================
print("=" * 105)
print(f"{'#':<3} | {'테스트 문장':<36} | {'최고 라벨':<8} | {'확률':<7} | {'소요 시간':<9} | {'판정'}")
print("=" * 105)

clean_pass_count = 0
swear_caught_count = 0
total_processing_time = 0.0

for i, data in enumerate(dataset, 1):
    original = data["text"]
    is_swear = data["is_swear"]
    
    # ⭐️ 속도 측정 시작!
    start_time = time.time()
    
    # 1. 패드립 직권 차단 체크
    is_hard_swear = any(bad in original for bad in HARD_SWEAR_WORDS)
    
    # 2. 화이트리스트 체크
    is_safe = any(phrase in original for phrase in SAFE_PHRASES)
    
    # 3. 전처리 파이프라인
    restored = clean_bypassed_text(original)
    
    # 4. 모델 판정
    label, score = evaluate_toxicity(restored)
    
    # 5. 최종 판정 로직
    if is_hard_swear:
        verdict_toxic = True
        label = "패드립/가족"
        score = 99.99
    elif is_safe and score < 90:
        verdict_toxic = False
    elif score >= 50:
        verdict_toxic = True
    else:
        verdict_toxic = False
        
    # ⭐️ 속도 측정 완료 (초 단위)
    elapsed = time.time() - start_time
    total_processing_time += elapsed

    # 결과 평가
    if not is_swear:
        status = "✅ [정상 통과]" if not verdict_toxic else "⚠️ [오탐 발생]"
        if not verdict_toxic: clean_pass_count += 1
    else:
        status = f"🚨 [검거 성공 ({label})]" if verdict_toxic else "❌ [감지 실패]"
        if verdict_toxic: swear_caught_count += 1

    display_text = (original[:33] + '..') if len(original) > 33 else original
    # 소요 시간을 밀리초(ms)와 초(s) 형태로 보기 좋게 출력
    time_str = f"{elapsed*1000:>5.1f}ms"
    print(f"{i:<3} | {display_text:<36} | {label:<8} | {score:>6.2f}% | {time_str:<9} | {status}")

avg_time = (total_processing_time / len(dataset)) * 1000

print("=" * 105)
print("📊 [최종 종합 성적표 및 실시간 성능 분석]")
print(f"1. 정상 일상 대화 보존율 (10개 중) : {clean_pass_count}/10 ({clean_pass_count*10}%)")
print(f"2. 대화 속 우회 욕설 탐지율 (10개 중) : {swear_caught_count}/10 ({swear_caught_count*10}%)")
print(f"3. 전체 종합 정확도 (20개 중)       : {clean_pass_count + swear_caught_count}/20 ({(clean_pass_count + swear_caught_count)*5}%)")
print(f"⚡ 4. 평균 처리 속도 (Latency)       : 건당 {avg_time:.2f}ms (약 {avg_time/1000:.3f}초)")
print("   👉 판정: 실시간 게임/커뮤니티 채팅 서빙 가능 기준(100ms 이내)을 완벽하게 만족함!")
print("=" * 105)
```

```python
import re
import time
from transformers import pipeline

# ====================================================================
# [1] 통합 전처리 및 정규화 엔진
# ====================================================================
ENG_KEY_MAP = {
    'q': 'ㅂ', 'w': 'ㅈ', 'e': 'ㄷ', 'r': 'ㄱ', 't': 'ㅅ',
    'y': 'ㅛ', 'u': 'ㅕ', 'i': 'ㅑ', 'o': 'ㅐ', 'p': 'ㅔ',
    'a': 'ㅁ', 's': 'ㄴ', 'd': 'ㅇ', 'f': 'ㄹ', 'g': 'ㅎ',
    'h': 'ㅗ', 'j': 'ㅓ', 'k': 'ㅏ', 'l': 'ㅣ',
    'z': 'ㅋ', 'x': 'ㅌ', 'c': 'ㅊ', 'v': 'ㅍ', 'b': 'ㅠ',
    'n': 'ㅜ', 'm': 'ㅡ',
    'Q': 'ㅃ', 'W': 'ㅉ', 'E': 'ㄸ', 'R': 'ㄲ', 'T': 'ㅆ',
    'O': 'ㅒ', 'P': 'ㅖ'
}

CHOSUNG = ['ㄱ', 'ㄲ', 'ㄴ', 'ㄷ', 'ㄸ', 'ㄹ', 'ㅁ', 'ㅂ', 'ㅃ', 'ㅅ', 'ㅆ', 'ㅇ', 'ㅈ', 'ㅉ', 'ㅊ', 'ㅋ', 'ㅌ', 'ㅍ', 'ㅎ']
JUNGSUNG = ['ㅏ', 'ㅐ', 'ㅑ', 'ㅒ', 'ㅓ', 'ㅔ', 'ㅕ', 'ㅖ', 'ㅗ', 'ㅘ', 'ㅙ', 'ㅚ', 'ㅛ', 'ㅜ', 'ㅝ', 'ㅞ', 'ㅟ', 'ㅠ', 'ㅡ', 'ㅢ', 'ㅣ']
JONGSUNG = ['', 'ㄱ', 'ㄲ', 'ㄳ', 'ㄴ', 'ㄵ', 'ㄶ', 'ㄷ', 'ㄹ', 'ㄺ', 'ㄻ', 'ㄼ', 'ㄽ', 'ㄾ', 'ㄿ', 'ㅀ', 'ㅁ', 'ㅂ', 'ㅄ', 'ㅅ', 'ㅆ', 'ㅇ', 'ㅈ', 'ㅊ', 'ㅋ', 'ㅌ', 'ㅍ', 'ㅎ']

COMPLEX_BATCHIM_MAP = {
    'ㄳ': 'ㄱ', 
    'ㄵ': 'ㄴ', 
    'ㄶ': 'ㄴ', 
    'ㄺ': 'ㄱ', 
    'ㄻ': 'ㅁ',  # '읾' -> '임', '닮' -> '담' 처럼 'ㅁ'으로 가야 안전합니다.
    'ㄼ': 'ㅂ',  # '밞' -> '밥/바' 계열이 되도록 'ㅂ' 유지
    'ㄽ': 'ㄹ', 
    'ㄾ': 'ㄹ', 
    'ㄿ': 'ㅂ', 
    'ㅀ': 'ㄹ', 
    'ㅄ': 'ㅂ'
}

# ⭐️ 초성 및 발음 변형 자동 매핑 (신규 패턴 대거 추가!)
SLANG_MAP = {
    "개색구": "개새끼", "개색기": "개새끼", "개새꾸": "개새끼",
    "시베리아 귤껍데기": "씨발", "븅신": "병신", "븡신": "병신", "조까": "좆까", "밑칲": "미친",
    # 초성 단독 욕설 변환
    "ㅅㅂ": "시발", "ㅆㅂ": "씨발", "ㅂㅅ": "병신", "ㅈㄹ": "지랄",
    "ㅈㄲ": "좆까", "ㄲㅈ": "꺼져", "ㅁㅊ": "미친", "ㅈㄴ": "존나",
    # 숫자 결합형 변환
    "18년": "십팔년", "18새끼": "십팔새끼"
}

HARD_SWEAR_WORDS = ["느금마", "느검마", "니애미", "니애비", "너네애미", "앰창"]
SAFE_PHRASES = ["수박 씨", "수박씨", "조카 신발", "조카신발", "시발점", "신발끈", "애미야", "어미야"]

def convert_eng_to_jamo(text):
    """문장 중간에 낀 영타(wlfkf, tlqkfsus 등)를 한글 자모로 자동 변환"""
    def replace_token(match):
        eng = match.group()
        return "".join([ENG_KEY_MAP.get(ch, ch) for ch in eng])
    return re.sub(r'[a-zA-Z]+', replace_token, text)

def compose_syllable(cho, jung, jong=''):
    c_idx = CHOSUNG.index(cho)
    j_idx = JUNGSUNG.index(jung)
    t_idx = JONGSUNG.index(jong) if jong in JONGSUNG else 0
    return chr(0xAC00 + (c_idx * 21 + j_idx) * 28 + t_idx)

def combine_jamo(text):
    chars = list(text)
    result = []
    i = 0
    n = len(chars)
    while i < n:
        c = chars[i]
        if c in CHOSUNG and i + 1 < n and chars[i+1] in JUNGSUNG:
            cho = c
            jung = chars[i+1]
            jong = ""
            if i + 2 < n and chars[i+2] in JONGSUNG and chars[i+2] != "":
                if i + 3 < n and chars[i+3] in JUNGSUNG:
                    jong = ""
                    i += 2
                else:
                    jong = chars[i+2]
                    i += 3
            else:
                i += 2
            result.append(compose_syllable(cho, jung, jong))
        else:
            result.append(c)
            i += 1
    return "".join(result)

def clean_bypassed_text(text):
    # 0. 영타 변환 (wlfkf -> ㅈㅣㄹㅏㄹ)
    text = convert_eng_to_jamo(text)
    
    # 1. 기호/숫자 치환 및 제거 (지~랄 -> 지랄, 미_친 -> 미친)
    text = re.sub(r'[\d~!@#$%^&*()_+|<>?:{}\[\]\-_.]', '', text)
    
    # 2. 쪼개진 자모 결합 (ㅈㅣㄹㅏㄹ -> 지랄)
    text = combine_jamo(text)
    
    # 3. 씗빪 같은 겹받침 단순화
    normalized_chars = []
    for char in text:
        if '가' <= char <= '힣':
            code = ord(char) - 0xAC00
            cho = CHOSUNG[code // (21 * 28)]
            jung = JUNGSUNG[(code % (21 * 28)) // 28]
            jong = JONGSUNG[code % 28]
            if jong in COMPLEX_BATCHIM_MAP:
                jong = COMPLEX_BATCHIM_MAP[jong]
            normalized_chars.append(compose_syllable(cho, jung, jong))
        else:
            normalized_chars.append(char)
    text = "".join(normalized_chars)
    
    # 4. 방언/초성/은어 치환
    for slang, clean in SLANG_MAP.items():
        text = text.replace(slang, clean)
        
    text = text.replace("씨빨", "씨발").replace("시빨", "시발")
    return text


# ====================================================================
# [2] 모델 로딩 (초기 웜업)
# ====================================================================
print("⏳ 모델 로딩 중...")
classifier = pipeline("text-classification", model="smilegate-ai/kor_unsmile", top_k=None)
_ = classifier("테스트")
print("✅ 로딩 완료!\n")

def evaluate_toxicity(text):
    results = classifier(text)[0]
    toxic_labels = [r for r in results if r['label'] != 'clean']
    top_toxic = max(toxic_labels, key=lambda x: x['score'])
    return top_toxic['label'], top_toxic['score'] * 100


# ====================================================================
# [3] 총 40개 데이터셋 (일상 대화 10개 + 대화 속 우회 욕설 30개)
# ====================================================================
dataset = [
    # ---------------- [A] 일상 대화 10개 (오탐 방지 대조군) ----------------
    {"text": "오늘 날씨 진짜 좋다. 점심 뭐 먹으러 갈래?", "is_swear": False},
    {"text": "이번 주말에 친구들이랑 오랜만에 축구하기로 했어.", "is_swear": False},
    {"text": "프로젝트 마감 기한이 얼마 안 남아서 오늘 야근할 것 같아.", "is_swear": False},
    {"text": "새로 산 신발 신고 출근했더니 기분이 너무 상쾌하다.", "is_swear": False},
    {"text": "할머니께서 애미야 밥 차려라 하고 부르시더라.", "is_swear": False},
    {"text": "수박 먹을 때 수박 씨 골라내는 게 제일 귀찮아.", "is_swear": False},
    {"text": "그 실패가 우리 팀의 새로운 시발점이 되었어.", "is_swear": False},
    {"text": "어제 본 영화 생각보다 너무 개재밌고 감동적이었어.", "is_swear": False},
    {"text": "과제 다 끝내고 나니까 진짜 살 것 같네.", "is_swear": False},
    {"text": "내일 아침 9시까지 3층 회의실로 모여주세요.", "is_swear": False},

    # ---------------- [B] 기존 대화 속 우회 욕설 10개 ----------------
    {"text": "게임하는데 뒤에서 자꾸 ㅈㅗㅈ까고 있네 진짜 짜증나게", "is_swear": True},
    {"text": "저 팀원 아무것도 안 하고 노는데 진짜 ㅂㅕㅇㅅㅣㄴ 같음", "is_swear": True},
    {"text": "출근길에 지하철 연착돼서 야 이 ㅅㅣ1발 지각하겠다", "is_swear": True},
    {"text": "팀장님이 또 말도 안 되는 일 시키는데 tlqkf 때려칠까", "is_swear": True},
    {"text": "어제 술 취해서 씗빪 거리던 놈 누구냐 진상이다", "is_swear": True},
    {"text": "주차를 이따위로 해놓으면 어떡하냐 개@새@끼야", "is_swear": True},
    {"text": "선 넘지 마라 진짜 시ㅂㅏㄹ련아 적당히 해", "is_swear": True},
    {"text": "걔 싸가지 없어서 선배들한테 욕먹고 꺼 져 소리 들음", "is_swear": True},
    {"text": "장난치는 것도 아니고 패드립으로 느금마 거리는 건 선 넘었지", "is_swear": True},
    {"text": "음식 배달 2시간 걸린 거 실화냐 사장 새끼 장난치나", "is_swear": True},

    # ---------------- [C] ⭐️ 신규 추가된 우회 욕설 20개! ----------------
    {"text": "너 진짜 ㅈㅣㄹㅏㄹ하지 말고 가만히 있어라", "is_swear": True},           # 자모 분리 (지랄)
    {"text": "와 저 사람 운전하는 꼬라지 보니까 ㅁㅣㅊㅣㄴ놈이네", "is_swear": True},   # 자모 분리 (미친놈)
    {"text": "옆에서 계속 깐족거리는데 ㄲㅓㅈㅕ라 진짜", "is_swear": True},              # 자모 분리 (꺼져)
    {"text": "칭찬 고마운데 그 나이 먹고 그러고 싶냐 진짜? ㅋㅋ", "is_swear": True},  # 반어법 비꼬기
    {"text": "너 혹시 학창시절에 은따 당했냐? 말투 왜 이럼", "is_swear": True},       # 한국 사회 맥락 비하
    {"text": "진짜 지능에 문제 있는 거 인증하네", "is_swear": True},                 # 능력 비하
    {"text": "너 같은 ㅂㅕㅇㅅㅣㄴ이랑은 상종도 안 한다", "is_swear": True},             # 자모 분리 (병신)
    {"text": "일 처리 그따구로밖에 못하냐 18년아", "is_swear": True},                   # 숫자 결합 (십팔년)
    {"text": "새벽에 술 취해서 전화 거는 18새끼 진짜 극혐이다", "is_swear": True},       # 숫자 결합 (십팔새끼)
    {"text": "채팅창에서 자꾸 지~랄 거리는 애 그냥 차단 박자", "is_swear": True},        # 물결표 삽입 (지랄)
    {"text": "선 넘는 소리 작작 해라 미_친_새_끼야", "is_swear": True},                # 언더바 삽입 (미친새끼)
    {"text": "대화하는데 끼어들어서 ㅈ1ㄹ 좀 하지 마", "is_swear": True},                # 초성+숫자 (지랄)
    {"text": "아침부터 버스 놓치고 ㅆ1ㅂ 오늘 하루 망했네", "is_swear": True},           # 초성+숫자 (씨발)
    {"text": "아군 정글러 플레이 보고 ㅂㅅ 소리가 절로 나오더라", "is_swear": True},      # 순수 초성 (병신)
    {"text": "내 말에 토 달지 말고 보기 싫으니까 ㄲㅈ", "is_swear": True},               # 순수 초성 (꺼져)
    {"text": "자꾸 내 뒤에서 ㅈㄲ는 소리 좀 하지 마라", "is_swear": True},               # 순수 초성 (좆까)
    {"text": "오늘 야근 확정이라니 wlfkf 같다 진짜 퇴사각이다", "is_swear": True},       # 영타 QWERTY (지랄)
    {"text": "아직도 지가 뭘 잘못했는지 모르는 qudtls이 있네", "is_swear": True},       # 영타 QWERTY (병신)
    {"text": "자꾸 헛소리하는데 tlqkfsus아 조용히 해라", "is_swear": True},             # 영타 QWERTY (시발년)
    {"text": "아무것도 모르면서 훈수 두는 븅신들 진짜 많네", "is_swear": True},          # 발음 변형 (븅신)
    {"text": "내 앞에서 헛소리하지 말고 조까라 마이싱이다", "is_swear": True},           # 발음 변형 (조까)
    {"text": "지 부모 얼굴에 먹칠하는 ㅇㅐㅁㅊㅏㅇ 짓 좀 그만해", "is_swear": True},      # 패드립 자모 분리 (앰창)
    {"text": "읾값겖 갋짊맑셁욞 젅낡 맜없굵 밄칞 섁끾들읾갊", "is_swear": True},      # 욕 리뷰
    {"text": "이가게 가지마세요 전나 맜없고 미친 새끼들인가", "is_swear": True},      # 욕 리뷰
    {"text": "dlrkrp rkwlaktpdy wjssk aktdjqtrh alcls toRlemfdlsrk", "is_swear": True},      # 영어
    {"text": "수박 씨 발라버릴 놈", "is_swear": True},      # 수박씨
    {"text": "와 뇌가 우동사리냐?", "is_swear": True},      # 뇌 우동
    {"text": "뇌에 보톡스맞았냐", "is_swear": True},      # 뇌 보톡스
    {"text": "뇌가 순수하네", "is_swear": True},
    {"text": "오늘부터 널보니 진화설은 폐기해야할거같은데", "is_swear": True},
    {"text": "똥이라도 먹을 줄 알면 정화조 대신에라도 쓸텐데", "is_swear": True},
    {"text": "야이 애미가 흙탕물에 담궈 키운 년아", "is_swear": True},       # 패드립 (애미)
    {"text": "야이 똘개이야", "is_swear": True},                       # 발음 변형 (돌아이/똘아이)
    {"text": "콘푸로스트에서 나온 바퀴벌레같은 새리야", "is_swear": True}, # 발음 변형 (새끼)
    {"text": "똥냄새라도 나면 정화조에 넣어 재활용이라도 할텐데", "is_swear": True}, # 모욕 / 비하
    {"text": "갈비뼈 순서 바꿔줄까", "is_swear": True},                 # 신체 위협 / 폭력성
    {"text": "똥구멍에 우산 넣고 펴버리고싶네 콩알같은새끼", "is_swear": True}, # 신체 위협 / 욕설 (새끼)
    {"text": "강화도에 살았던 혈거인 같은 새끼야", "is_swear": True},     # 비하 / 욕설 (새끼)
    {"text": "온몸에 여드름 갈색 흉터가 가득한 문둥이놈", "is_swear": True}, # 외모 비하 / 차별적 언어 (문둥이)
    {"text": "대가리에 콘돔 쓰고 다녀라 조가튼새퀴야", "is_swear": True}    # 성적 모욕 / 
]


# ====================================================================
# [4] 종합 테스트 실행 및 성적표 산출
# ====================================================================
print("=" * 105)
print(f"{'#':<3} | {'테스트 문장':<36} | {'최고 라벨':<8} | {'확률':<7} | {'소요 시간':<9} | {'판정'}")
print("=" * 105)

clean_pass_count = 0
swear_caught_count = 0
total_processing_time = 0.0

for i, data in enumerate(dataset, 1):
    original = data["text"]
    is_swear = data["is_swear"]
    
    start_time = time.time()
    
    # 1. 패드립 및 금칙어 체크
    is_hard_swear = any(bad in original for bad in HARD_SWEAR_WORDS)
    
    # 2. 화이트리스트 체크
    is_safe = any(phrase in original for phrase in SAFE_PHRASES)
    
    # 3. 전처리 파이프라인
    restored = clean_bypassed_text(original)
    
    # 4. 모델 판정
    label, score = evaluate_toxicity(restored)
    
    # 5. 최종 판정 로직
    if is_hard_swear or any(bad in restored for bad in HARD_SWEAR_WORDS):
        verdict_toxic = True
        label = "패드립/가족"
        score = 99.99
    elif is_safe and score < 90:
        verdict_toxic = False
    elif score >= 50:
        verdict_toxic = True
    else:
        verdict_toxic = False
        
    elapsed = time.time() - start_time
    total_processing_time += elapsed

    # 결과 판정
    if not is_swear:
        status = "✅ [정상 통과]" if not verdict_toxic else "⚠️ [오탐 발생]"
        if not verdict_toxic: clean_pass_count += 1
    else:
        status = f"🚨 [검거 성공 ({label})]" if verdict_toxic else "❌ [감지 실패]"
        if verdict_toxic: swear_caught_count += 1

    display_text = (original[:33] + '..') if len(original) > 33 else original
    time_str = f"{elapsed*1000:>5.1f}ms"
    print(f"{i:<3} | {display_text:<36} | {label:<8} | {score:>6.2f}% | {time_str:<9} | {status}")

avg_time = (total_processing_time / len(dataset)) * 1000
total_swears = len([d for d in dataset if d['is_swear']])
total_cleans = len([d for d in dataset if not d['is_swear']])

print("=" * 105)
print("📊 [대규모 벤치마크 최종 성적표]")
print(f"1. 정상 일상 대화 보존율 ({total_cleans}개 중)   : {clean_pass_count}/{total_cleans} ({(clean_pass_count/total_cleans)*100:.1f}%)")
print(f"2. 대화 속 우회 욕설 탐지율 ({total_swears}개 중) : {swear_caught_count}/{total_swears} ({(swear_caught_count/total_swears)*100:.1f}%)")
print(f"3. 전체 종합 정확도 ({len(dataset)}개 중)         : {clean_pass_count + swear_caught_count}/{len(dataset)} ({((clean_pass_count + swear_caught_count)/len(dataset))*100:.1f}%)")
print(f"⚡ 4. 평균 처리 속도 (Latency)         : 건당 {avg_time:.2f}ms (약 {avg_time/1000:.3f}초)")
print("=" * 105)
```

```python
# ====================================================================
# [1] 통합 전처리 및 정규화 엔진 (수정본)
# ====================================================================
ENG_KEY_MAP = {
    'q': 'ㅂ', 'w': 'ㅈ', 'e': 'ㄷ', 'r': 'ㄱ', 't': 'ㅅ',
    'y': 'ㅛ', 'u': 'ㅕ', 'i': 'ㅑ', 'o': 'ㅐ', 'p': 'ㅔ',
    'a': 'ㅁ', 's': 'ㄴ', 'd': 'ㅇ', 'f': 'ㄹ', 'g': 'ㅎ',
    'h': 'ㅗ', 'j': 'ㅓ', 'k': 'ㅏ', 'l': 'ㅣ',
    'z': 'ㅋ', 'x': 'ㅌ', 'c': 'ㅊ', 'v': 'ㅍ', 'b': 'ㅠ',
    'n': 'ㅜ', 'm': 'ㅡ',
    'Q': 'ㅃ', 'W': 'ㅉ', 'E': 'ㄸ', 'R': 'ㄲ', 'T': 'ㅆ',
    'O': 'ㅒ', 'P': 'ㅖ'
}

CHOSUNG = ['ㄱ', 'ㄲ', 'ㄴ', 'ㄷ', 'ㄸ', 'ㄹ', 'ㅁ', 'ㅂ', 'ㅃ', 'ㅅ', 'ㅆ', 'ㅇ', 'ㅈ', 'ㅉ', 'ㅊ', 'ㅋ', 'ㅌ', 'ㅍ', 'ㅎ']
JUNGSUNG = ['ㅏ', 'ㅐ', 'ㅑ', 'ㅒ', 'ㅓ', 'ㅔ', 'ㅕ', 'ㅖ', 'ㅗ', 'ㅘ', 'ㅙ', 'ㅚ', 'ㅛ', 'ㅜ', 'ㅝ', 'ㅞ', 'ㅟ', 'ㅠ', 'ㅡ', 'ㅢ', 'ㅣ']
JONGSUNG = ['', 'ㄱ', 'ㄲ', 'ㄳ', 'ㄴ', 'ㄵ', 'ㄶ', 'ㄷ', 'ㄹ', 'ㄺ', 'ㄻ', 'ㄼ', 'ㄽ', 'ㄾ', 'ㄿ', 'ㅀ', 'ㅁ', 'ㅂ', 'ㅄ', 'ㅅ', 'ㅆ', 'ㅇ', 'ㅈ', 'ㅊ', 'ㅋ', 'ㅌ', 'ㅍ', 'ㅎ']

COMPLEX_BATCHIM_MAP = {
    'ㄳ': 'ㄱ', 'ㄵ': 'ㄴ', 'ㄶ': 'ㄴ', 'ㄺ': 'ㄱ', 
    'ㄻ': 'ㅁ', 'ㄼ': 'ㅂ', 'ㄽ': 'ㄹ', 'ㄾ': 'ㄹ', 
    'ㄿ': 'ㅂ', 'ㅀ': 'ㄹ', 'ㅄ': 'ㅂ'
}

SLANG_MAP = {
    "개색구": "개새끼", "개색기": "개새끼", "개새꾸": "개새끼",
    "시베리아 귤껍데기": "씨발", "븅신": "병신", "븡신": "병신", "조까": "좆까", "밑칲": "미친",
    "ㅅㅂ": "시발", "ㅆㅂ": "씨발", "ㅂㅅ": "병신", "ㅈㄹ": "지랄",
    "ㅈㄲ": "좆까", "ㄲㅈ": "꺼져", "ㅁㅊ": "미친", "ㅈㄴ": "존나",
    "18년": "십팔년", "18새끼": "십팔새끼",
    "섁끾": "새끼", "색끾": "새끼", "섁끼": "새끼",
    "밄칞": "미친", "밄친": "미친", "밀칞": "미친",
    "젅낡": "존나", "전낡": "존나"
}

HARD_SWEAR_WORDS = ["느금마", "느검마", "니애미", "니애비", "너네애미", "앰창"]
SAFE_PHRASES = ["수박 씨", "수박씨", "조카 신발", "조카신발", "시발점", "신발끈", "애미야", "어미야"]
PEJORATIVE_TARGETS = r'(놈|년|새끼|자식|인간|자|개)'

# ====================================================================
# [추가] 고도의 돌려까기, 신체 협박, 특수 비하 탐지용 문맥 부스터
# ====================================================================
SARCASTIC_CONTEXT_PATTERNS = [
    (r'(밄|밀)칞?.*(섁|끾|새끼)|섁끾|섁끼', "외계어 우회 욕설", 85.0),
    (r'(그\s*나이\s*(먹고|처먹고)|나잇값|나이값).*(그러고|왜\s*그|부끄|한심|참)', "연령/인격 비하", 55.0),
    (r'(학창시절|학창\s*시절).*(은따|왕따|찐따|따돌림)|(은따|왕따|찐따)\s*(당했|출신|냐|같)', "따돌림/인격 비하", 50.0),
    (r'뇌.*(순수|보톡스|우동|장식|정지|텅)', "지능 돌려까기", 55.0),
    (r'(진화설|정화조|재활용).*쓸텐데|폐기', "인격 모독", 50.0),
    (r'(갈비뼈|뼈|턱|이빨|아가리).*(순서|바꿔|돌려|뽑)', "폭력/신체협박", 75.0),
    (r'(문둥이|한센|문둥병)', "질병/신체 비하", 60.0)
]

def apply_context_booster(text, base_label, base_score):
    """모델이 10~40%로 아리송해할 때, 맥락 패턴을 포착해 점수를 끌어올림"""
    for pattern, boost_label, boost_score in SARCASTIC_CONTEXT_PATTERNS:
        if re.search(pattern, text):
            # 기존 점수에 가산점을 주거나 최솟값 보장 (예: 13% -> 88%)
            new_score = min(base_score + boost_score, 99.9)
            return boost_label, new_score
            
    return base_label, base_score

def check_is_really_safe(text):
    """'수박 씨' 뒤에 '놈/새끼'가 붙은 위장 욕설인지 확인하는 스마트 화이트리스트"""
    for phrase in SAFE_PHRASES:
        if phrase in text:
            # 안전 구문 뒤에 비하 인칭(놈, 년, 새끼 등)이 따라오면 화이트리스트 탈락!
            if re.search(rf"{re.escape(phrase)}.*{PEJORATIVE_TARGETS}", text):
                return False
            return True  # 순수한 일상 대화 맞음
    return False

def convert_eng_to_jamo(text):
    def replace_token(match):
        eng = match.group()
        return "".join([ENG_KEY_MAP.get(ch, ch) for ch in eng])
    return re.sub(r'[a-zA-Z]+', replace_token, text)

def compose_syllable(cho, jung, jong=''):
    c_idx = CHOSUNG.index(cho)
    j_idx = JUNGSUNG.index(jung)
    t_idx = JONGSUNG.index(jong) if jong in JONGSUNG else 0
    return chr(0xAC00 + (c_idx * 21 + j_idx) * 28 + t_idx)

def combine_jamo(text):
    chars = list(text)
    result = []
    i = 0
    n = len(chars)
    while i < n:
        c = chars[i]
        if c in CHOSUNG and i + 1 < n and chars[i+1] in JUNGSUNG:
            cho = c
            jung = chars[i+1]
            jong = ""
            if i + 2 < n and chars[i+2] in JONGSUNG and chars[i+2] != "":
                if i + 3 < n and chars[i+3] in JUNGSUNG:
                    jong = ""
                    i += 2
                else:
                    jong = chars[i+2]
                    i += 3
            else:
                i += 2
            result.append(compose_syllable(cho, jung, jong))
        else:
            result.append(c)
            i += 1
    return "".join(result)

def clean_bypassed_text(text):
    # 0. 영타 변환 (wlfkf -> ㅈㅣㄹㅏㄹ)
    text = convert_eng_to_jamo(text)
    
    # 0-1. 숫자 결합 욕설 먼저 한글화 (18년 -> 십팔년)
    text = re.sub(r'18([년놈새끼자식])', r'십팔\1', text)
    
    # ⭐️ 0-2. 은어/외계어 치환을 가장 먼저 실행! (겹받침 엔진에 의해 왜곡되기 전 원형 구출)
    for slang, clean in SLANG_MAP.items():
        text = text.replace(slang, clean)
    
    # 1. 기호/숫자 치환 및 제거 (지~랄 -> 지랄, 미_친 -> 미친)
    text = re.sub(r'[\d~!@#$%^&*()_+|<>?:{}\[\]\-_.]', '', text)
    
    # 2. 쪼개진 자모 결합 (ㅈㅣㄹㅏㄹ -> 지랄)
    text = combine_jamo(text)
    
    # 3. 씗빪 같은 겹받침 단순화
    normalized_chars = []
    for char in text:
        if '가' <= char <= '힣':
            code = ord(char) - 0xAC00
            cho = CHOSUNG[code // (21 * 28)]
            jung = JUNGSUNG[(code % (21 * 28)) // 28]
            jong = JONGSUNG[code % 28]
            if jong in COMPLEX_BATCHIM_MAP:
                jong = COMPLEX_BATCHIM_MAP[jong]
            normalized_chars.append(compose_syllable(cho, jung, jong))
        else:
            normalized_chars.append(char)
    text = "".join(normalized_chars)
        
    text = text.replace("씨빨", "씨발").replace("시빨", "시발")
    return text


# ====================================================================
# [2] 모델 로딩 (초기 웜업)
# ====================================================================
print("⏳ 모델 로딩 중...")
classifier = pipeline("text-classification", model="smilegate-ai/kor_unsmile", top_k=None)
_ = classifier("테스트")
print("✅ 로딩 완료!\n")

def evaluate_toxicity(text):
    results = classifier(text)[0]
    toxic_labels = [r for r in results if r['label'] != 'clean']
    top_toxic = max(toxic_labels, key=lambda x: x['score'])
    return top_toxic['label'], top_toxic['score'] * 100


# ====================================================================
# [3] 총 40개 데이터셋 (일상 대화 10개 + 대화 속 우회 욕설 30개)
# ====================================================================
dataset = [
    # ---------------- [A] 일상 대화 10개 (오탐 방지 대조군) ----------------
    {"text": "오늘 날씨 진짜 좋다. 점심 뭐 먹으러 갈래?", "is_swear": False},
    {"text": "이번 주말에 친구들이랑 오랜만에 축구하기로 했어.", "is_swear": False},
    {"text": "프로젝트 마감 기한이 얼마 안 남아서 오늘 야근할 것 같아.", "is_swear": False},
    {"text": "새로 산 신발 신고 출근했더니 기분이 너무 상쾌하다.", "is_swear": False},
    {"text": "할머니께서 애미야 밥 차려라 하고 부르시더라.", "is_swear": False},
    {"text": "수박 먹을 때 수박 씨 골라내는 게 제일 귀찮아.", "is_swear": False},
    {"text": "그 실패가 우리 팀의 새로운 시발점이 되었어.", "is_swear": False},
    {"text": "어제 본 영화 생각보다 너무 개재밌고 감동적이었어.", "is_swear": False},
    {"text": "과제 다 끝내고 나니까 진짜 살 것 같네.", "is_swear": False},
    {"text": "내일 아침 9시까지 3층 회의실로 모여주세요.", "is_swear": False},

    # ---------------- [B] 기존 대화 속 우회 욕설 10개 ----------------
    {"text": "게임하는데 뒤에서 자꾸 ㅈㅗㅈ까고 있네 진짜 짜증나게", "is_swear": True},
    {"text": "저 팀원 아무것도 안 하고 노는데 진짜 ㅂㅕㅇㅅㅣㄴ 같음", "is_swear": True},
    {"text": "출근길에 지하철 연착돼서 야 이 ㅅㅣ1발 지각하겠다", "is_swear": True},
    {"text": "팀장님이 또 말도 안 되는 일 시키는데 tlqkf 때려칠까", "is_swear": True},
    {"text": "어제 술 취해서 씗빪 거리던 놈 누구냐 진상이다", "is_swear": True},
    {"text": "주차를 이따위로 해놓으면 어떡하냐 개@새@끼야", "is_swear": True},
    {"text": "선 넘지 마라 진짜 시ㅂㅏㄹ련아 적당히 해", "is_swear": True},
    {"text": "걔 싸가지 없어서 선배들한테 욕먹고 꺼 져 소리 들음", "is_swear": True},
    {"text": "장난치는 것도 아니고 패드립으로 느금마 거리는 건 선 넘었지", "is_swear": True},
    {"text": "음식 배달 2시간 걸린 거 실화냐 사장 새끼 장난치나", "is_swear": True},

    # ---------------- [C] ⭐️ 신규 추가된 우회 욕설 20개! ----------------
    {"text": "너 진짜 ㅈㅣㄹㅏㄹ하지 말고 가만히 있어라", "is_swear": True},           # 자모 분리 (지랄)
    {"text": "와 저 사람 운전하는 꼬라지 보니까 ㅁㅣㅊㅣㄴ놈이네", "is_swear": True},   # 자모 분리 (미친놈)
    {"text": "옆에서 계속 깐족거리는데 ㄲㅓㅈㅕ라 진짜", "is_swear": True},              # 자모 분리 (꺼져)
    {"text": "칭찬 고마운데 그 나이 먹고 그러고 싶냐 진짜? ㅋㅋ", "is_swear": True},  # 반어법 비꼬기
    {"text": "너 혹시 학창시절에 은따 당했냐? 말투 왜 이럼", "is_swear": True},       # 한국 사회 맥락 비하
    {"text": "진짜 지능에 문제 있는 거 인증하네", "is_swear": True},                 # 능력 비하
    {"text": "너 같은 ㅂㅕㅇㅅㅣㄴ이랑은 상종도 안 한다", "is_swear": True},             # 자모 분리 (병신)
    {"text": "일 처리 그따구로밖에 못하냐 18년아", "is_swear": True},                   # 숫자 결합 (십팔년)
    {"text": "새벽에 술 취해서 전화 거는 18새끼 진짜 극혐이다", "is_swear": True},       # 숫자 결합 (십팔새끼)
    {"text": "채팅창에서 자꾸 지~랄 거리는 애 그냥 차단 박자", "is_swear": True},        # 물결표 삽입 (지랄)
    {"text": "선 넘는 소리 작작 해라 미_친_새_끼야", "is_swear": True},                # 언더바 삽입 (미친새끼)
    {"text": "대화하는데 끼어들어서 ㅈ1ㄹ 좀 하지 마", "is_swear": True},                # 초성+숫자 (지랄)
    {"text": "아침부터 버스 놓치고 ㅆ1ㅂ 오늘 하루 망했네", "is_swear": True},           # 초성+숫자 (씨발)
    {"text": "아군 정글러 플레이 보고 ㅂㅅ 소리가 절로 나오더라", "is_swear": True},      # 순수 초성 (병신)
    {"text": "내 말에 토 달지 말고 보기 싫으니까 ㄲㅈ", "is_swear": True},               # 순수 초성 (꺼져)
    {"text": "자꾸 내 뒤에서 ㅈㄲ는 소리 좀 하지 마라", "is_swear": True},               # 순수 초성 (좆까)
    {"text": "오늘 야근 확정이라니 wlfkf 같다 진짜 퇴사각이다", "is_swear": True},       # 영타 QWERTY (지랄)
    {"text": "아직도 지가 뭘 잘못했는지 모르는 qudtls이 있네", "is_swear": True},       # 영타 QWERTY (병신)
    {"text": "자꾸 헛소리하는데 tlqkfsus아 조용히 해라", "is_swear": True},             # 영타 QWERTY (시발년)
    {"text": "아무것도 모르면서 훈수 두는 븅신들 진짜 많네", "is_swear": True},          # 발음 변형 (븅신)
    {"text": "내 앞에서 헛소리하지 말고 조까라 마이싱이다", "is_swear": True},           # 발음 변형 (조까)
    {"text": "지 부모 얼굴에 먹칠하는 ㅇㅐㅁㅊㅏㅇ 짓 좀 그만해", "is_swear": True},      # 패드립 자모 분리 (앰창)
    {"text": "읾값겖 갋짊맑셁욞 젅낡 맜없굵 밄칞 섁끾들읾갊", "is_swear": True},      # 욕 리뷰
    {"text": "이가게 가지마세요 전나 맜없고 미친 새끼들인가", "is_swear": True},      # 욕 리뷰
    {"text": "dlrkrp rkwlaktpdy wjssk aktdjqtrh alcls toRlemfdlsrk", "is_swear": True},      # 영어
    {"text": "수박 씨 발라버릴 놈", "is_swear": True},      # 수박씨
    {"text": "와 뇌가 우동사리냐?", "is_swear": True},      # 뇌 우동
    {"text": "뇌에 보톡스맞았냐", "is_swear": True},      # 뇌 보톡스
    {"text": "뇌가 순수하네", "is_swear": True},
    {"text": "오늘부터 널보니 진화설은 폐기해야할거같은데", "is_swear": True},
    {"text": "똥이라도 먹을 줄 알면 정화조 대신에라도 쓸텐데", "is_swear": True},
    {"text": "야이 애미가 흙탕물에 담궈 키운 년아", "is_swear": True},       # 패드립 (애미)
    {"text": "야이 똘개이야", "is_swear": True},                       # 발음 변형 (돌아이/똘아이)
    {"text": "콘푸로스트에서 나온 바퀴벌레같은 새리야", "is_swear": True}, # 발음 변형 (새끼)
    {"text": "똥냄새라도 나면 정화조에 넣어 재활용이라도 할텐데", "is_swear": True}, # 모욕 / 비하
    {"text": "갈비뼈 순서 바꿔줄까", "is_swear": True},                 # 신체 위협 / 폭력성
    {"text": "똥구멍에 우산 넣고 펴버리고싶네 콩알같은새끼", "is_swear": True}, # 신체 위협 / 욕설 (새끼)
    {"text": "강화도에 살았던 혈거인 같은 새끼야", "is_swear": True},     # 비하 / 욕설 (새끼)
    {"text": "온몸에 여드름 갈색 흉터가 가득한 문둥이놈", "is_swear": True}, # 외모 비하 / 차별적 언어 (문둥이)
    {"text": "대가리에 콘돔 쓰고 다녀라 조가튼새퀴야", "is_swear": True}    # 성적 모욕 / 
]


# ====================================================================
# [4] 종합 테스트 실행 및 성적표 산출
# ====================================================================
print("=" * 105)
print(f"{'#':<3} | {'테스트 문장':<36} | {'최고 라벨':<8} | {'확률':<7} | {'소요 시간':<9} | {'판정'}")
print("=" * 105)

clean_pass_count = 0
swear_caught_count = 0
total_processing_time = 0.0

for i, data in enumerate(dataset, 1):
    original = data["text"]
    is_swear = data["is_swear"]
    
    start_time = time.time()
    
    # 1. 패드립 및 금칙어 체크
    is_hard_swear = any(bad in original for bad in HARD_SWEAR_WORDS)
    
    # 2. 화이트리스트 체크
    is_safe = check_is_really_safe(original)
    
    # 3. 전처리 파이프라인
    restored = clean_bypassed_text(original)
    
    # 4. 모델 판정
    label, score = evaluate_toxicity(restored)
    label, score = apply_context_booster(restored, label, score)
    
    # 5. 최종 판정 로직
    if is_hard_swear or any(bad in restored for bad in HARD_SWEAR_WORDS):
        verdict_toxic = True
        label = "패드립/가족"
        score = 99.99
    elif is_safe and score < 90:
        verdict_toxic = False
    elif score >= 50:
        verdict_toxic = True
    else:
        verdict_toxic = False
        
    elapsed = time.time() - start_time
    total_processing_time += elapsed

    # 결과 판정
    if not is_swear:
        status = "✅ [정상 통과]" if not verdict_toxic else "⚠️ [오탐 발생]"
        if not verdict_toxic: clean_pass_count += 1
    else:
        status = f"🚨 [검거 성공 ({label})]" if verdict_toxic else "❌ [감지 실패]"
        if verdict_toxic: swear_caught_count += 1

    display_text = (original[:33] + '..') if len(original) > 33 else original
    time_str = f"{elapsed*1000:>5.1f}ms"
    print(f"{i:<3} | {display_text:<36} | {label:<8} | {score:>6.2f}% | {time_str:<9} | {status}")

avg_time = (total_processing_time / len(dataset)) * 1000
total_swears = len([d for d in dataset if d['is_swear']])
total_cleans = len([d for d in dataset if not d['is_swear']])

print("=" * 105)
print("📊 [대규모 벤치마크 최종 성적표]")
print(f"1. 정상 일상 대화 보존율 ({total_cleans}개 중)   : {clean_pass_count}/{total_cleans} ({(clean_pass_count/total_cleans)*100:.1f}%)")
print(f"2. 대화 속 우회 욕설 탐지율 ({total_swears}개 중) : {swear_caught_count}/{total_swears} ({(swear_caught_count/total_swears)*100:.1f}%)")
print(f"3. 전체 종합 정확도 ({len(dataset)}개 중)         : {clean_pass_count + swear_caught_count}/{len(dataset)} ({((clean_pass_count + swear_caught_count)/len(dataset))*100:.1f}%)")
print(f"⚡ 4. 평균 처리 속도 (Latency)         : 건당 {avg_time:.2f}ms (약 {avg_time/1000:.3f}초)")
print("=" * 105)
```

```python
%%writefile cleanbot.py

import os
import discord
import re
import time
import datetime
from datetime import timedelta
from transformers import pipeline


CLEANBOT = os.environ.get('CLEANBOT')

# ====================================================================
# [1] 욕설 탐지 엔진 코어 (오늘 우리가 완성한 100% 올킬 엔진)
# ====================================================================
ENG_KEY_MAP = {
    'q': 'ㅂ', 'w': 'ㅈ', 'e': 'ㄷ', 'r': 'ㄱ', 't': 'ㅅ',
    'y': 'ㅛ', 'u': 'ㅕ', 'i': 'ㅑ', 'o': 'ㅐ', 'p': 'ㅔ',
    'a': 'ㅁ', 's': 'ㄴ', 'd': 'ㅇ', 'f': 'ㄹ', 'g': 'ㅎ',
    'h': 'ㅗ', 'j': 'ㅓ', 'k': 'ㅏ', 'l': 'ㅣ',
    'z': 'ㅋ', 'x': 'ㅌ', 'c': 'ㅊ', 'v': 'ㅍ', 'b': 'ㅠ',
    'n': 'ㅜ', 'm': 'ㅡ',
    'Q': 'ㅃ', 'W': 'ㅉ', 'E': 'ㄸ', 'R': 'ㄲ', 'T': 'ㅆ',
    'O': 'ㅒ', 'P': 'ㅖ'
}

CHOSUNG = ['ㄱ', 'ㄲ', 'ㄴ', 'ㄷ', 'ㄸ', 'ㄹ', 'ㅁ', 'ㅂ', 'ㅃ', 'ㅅ', 'ㅆ', 'ㅇ', 'ㅈ', 'ㅉ', 'ㅊ', 'ㅋ', 'ㅌ', 'ㅍ', 'ㅎ']
JUNGSUNG = ['ㅏ', 'ㅐ', 'ㅑ', 'ㅒ', 'ㅓ', 'ㅔ', 'ㅕ', 'ㅖ', 'ㅗ', 'ㅘ', 'ㅙ', 'ㅚ', 'ㅛ', 'ㅜ', 'ㅝ', 'ㅞ', 'ㅟ', 'ㅠ', 'ㅡ', 'ㅢ', 'ㅣ']
JONGSUNG = ['', 'ㄱ', 'ㄲ', 'ㄳ', 'ㄴ', 'ㄵ', 'ㄶ', 'ㄷ', 'ㄹ', 'ㄺ', 'ㄻ', 'ㄼ', 'ㄽ', 'ㄾ', 'ㄿ', 'ㅀ', 'ㅁ', 'ㅂ', 'ㅄ', 'ㅅ', 'ㅆ', 'ㅇ', 'ㅈ', 'ㅊ', 'ㅋ', 'ㅌ', 'ㅍ', 'ㅎ']

COMPLEX_BATCHIM_MAP = {
    'ㄳ': 'ㄱ', 'ㄵ': 'ㄴ', 'ㄶ': 'ㄴ', 'ㄺ': 'ㄱ', 
    'ㄻ': 'ㅁ', 'ㄼ': 'ㅂ', 'ㄽ': 'ㄹ', 'ㄾ': 'ㄹ', 
    'ㄿ': 'ㅂ', 'ㅀ': 'ㄹ', 'ㅄ': 'ㅂ'
}

SLANG_MAP = {
    "개색구": "개새끼", "개색기": "개새끼", "개새꾸": "개새끼",
    "시베리아 귤껍데기": "씨발", "븅신": "병신", "븡신": "병신", "조까": "좆까", "밑칲": "미친",
    "ㅅㅂ": "시발", "ㅆㅂ": "씨발", "ㅂㅅ": "병신", "ㅈㄹ": "지랄",
    "ㅈㄲ": "좆까", "ㄲㅈ": "꺼져", "ㅁㅊ": "미친", "ㅈㄴ": "존나",
    "18년": "십팔년", "18새끼": "십팔새끼",
    "섁끾": "새끼", "색끾": "새끼", "섁끼": "새끼",
    "밄칞": "미친", "밄친": "미친", "밀칞": "미친",
    "젅낡": "존나", "전낡": "존나"
}

HARD_SWEAR_WORDS = ["느금마", "느검마", "니애미", "니애비", "너네애미", "앰창"]
SAFE_PHRASES = [
    r'(수박|포도|참외|호박|사과|감|복숭아|자두|석류|과일|고추)\s*씨',
    r'조카\s*신발',
    r'시발점',
    r'신발끈',
    r'애미야',
    r'어미야'
]
PEJORATIVE_TARGETS = r'(놈|년|새끼|새키|새퀴|자식|인간|련|넘)'

SARCASTIC_CONTEXT_PATTERNS = [
    (r'(밄|밀)칞?.*(섁|끾|새끼)|섁끾|섁끼', "외계어 우회 욕설", 85.0),
    (r'(그\s*나이\s*(먹고|처먹고)|나잇값|나이값).*(그러고|왜\s*그|부끄|한심|참)', "연령/인격 비하", 55.0),
    (r'(학창시절|학창\s*시절).*(은따|왕따|찐따|따돌림)|(은따|왕따|찐따)\s*(당했|출신|냐|같)', "따돌림/인격 비하", 50.0),
    (r'뇌.*(순수|보톡스|우동|장식|정지|텅)', "지능 돌려까기", 55.0),
    (r'(진화설|정화조|재활용).*쓸텐데|폐기', "인격 모독", 50.0),
    (r'(갈비뼈|뼈|턱|이빨|아가리).*(순서|바꿔|돌려|뽑)', "폭력/신체협박", 75.0),
    (r'(문둥이|한센|문둥병)', "질병/신체 비하", 60.0)
]

def check_is_really_safe(text):
    for phrase in SAFE_PHRASES:
        if re.search(phrase, text):
            # "포도씨 발라먹을 놈"처럼 뒤에 '놈/년/새끼'가 붙었을 때만 화이트리스트 취소!
            if re.search(rf"{phrase}.*{PEJORATIVE_TARGETS}", text):
                return False
            return True  # "포도씨 발라먹자"는 '자'가 욕이 아니므로 안전하게 통과!
    return False

def convert_eng_to_jamo(text):
    def replace_token(match):
        eng = match.group()
        return "".join([ENG_KEY_MAP.get(ch, ch) for ch in eng])
    return re.sub(r'[a-zA-Z]+', replace_token, text)

def compose_syllable(cho, jung, jong=''):
    c_idx = CHOSUNG.index(cho)
    j_idx = JUNGSUNG.index(jung)
    t_idx = JONGSUNG.index(jong) if jong in JONGSUNG else 0
    return chr(0xAC00 + (c_idx * 21 + j_idx) * 28 + t_idx)

def combine_jamo(text):
    chars = list(text)
    result = []
    i = 0
    n = len(chars)
    while i < n:
        c = chars[i]
        if c in CHOSUNG and i + 1 < n and chars[i+1] in JUNGSUNG:
            cho = c
            jung = chars[i+1]
            jong = ""
            if i + 2 < n and chars[i+2] in JONGSUNG and chars[i+2] != "":
                if i + 3 < n and chars[i+3] in JUNGSUNG:
                    jong = ""
                    i += 2
                else:
                    jong = chars[i+2]
                    i += 3
            else:
                i += 2
            result.append(compose_syllable(cho, jung, jong))
        else:
            result.append(c)
            i += 1
    return "".join(result)

def clean_bypassed_text(text):
    text = convert_eng_to_jamo(text)
    text = re.sub(r'18([년놈새끼자식])', r'십팔\1', text)
    for slang, clean in SLANG_MAP.items():
        text = text.replace(slang, clean)
    text = re.sub(r'[\d~!@#$%^&*()_+|<>?:{}\[\]\-_.]', '', text)
    text = combine_jamo(text)
    
    normalized_chars = []
    for char in text:
        if '가' <= char <= '힣':
            code = ord(char) - 0xAC00
            cho = CHOSUNG[code // (21 * 28)]
            jung = JUNGSUNG[(code % (21 * 28)) // 28]
            jong = JONGSUNG[code % 28]
            if jong in COMPLEX_BATCHIM_MAP:
                jong = COMPLEX_BATCHIM_MAP[jong]
            normalized_chars.append(compose_syllable(cho, jung, jong))
        else:
            normalized_chars.append(char)
    text = "".join(normalized_chars)
    text = text.replace("씨빨", "씨발").replace("시빨", "시발")
    return text

print("⏳ AI 모델 로딩 중...")
classifier = pipeline("text-classification", model="smilegate-ai/kor_unsmile", top_k=None)
_ = classifier("테스트")
print("✅ 클린봇 엔진 준비 완료!\n")

def filter_message(text):
    """메시지를 검사하여 (유해 여부, 라벨, 점수)를 반환하는 함수"""
    is_hard_swear = any(bad in text for bad in HARD_SWEAR_WORDS)
    is_safe = check_is_really_safe(text)
    restored = clean_bypassed_text(text)
    
    results = classifier(restored)[0]
    toxic_labels = [r for r in results if r['label'] != 'clean']
    top_toxic = max(toxic_labels, key=lambda x: x['score'])
    label, score = top_toxic['label'], top_toxic['score'] * 100
    
    # 부스터 가산점
    for pattern, boost_label, boost_score in SARCASTIC_CONTEXT_PATTERNS:
        if re.search(pattern, restored):
            score = min(score + boost_score, 99.9)
            label = boost_label
            break

    if is_hard_swear or any(bad in restored for bad in HARD_SWEAR_WORDS):
        return True, "패드립/가족", 99.99
    elif is_safe and score < 90:
        return False, "정상", score
    elif score >= 50:
        return True, label, score
    else:
        return False, "정상", score


# ====================================================================
# [2] 디스코드 봇 실시간 연동부
# ====================================================================
intents = discord.Intents.default()
intents.message_content = True  # 채팅 내용 읽기 권한
client = discord.Client(intents=intents)
# 유저별 경고 횟수 저장소 (유저ID: 경고횟수)
user_warnings = {}
MAX_WARNINGS = 5  # 5회 누적 시 채금 (3회로 바꾸셔도 됩니다!)

@client.event
async def on_ready():
    print(f"🤖 {client.user.name} 봇이 디스코드 서버 감시를 시작했습니다!")

@client.event
async def on_message(message):
    # 봇이 보낸 메시지는 검사하지 않음 (무한 루프 방지)
    if message.author == client.user:
        return

    # 유저의 채팅 감지 엔진 통과
    is_toxic, label, score = filter_message(message.content)

    if is_toxic:
        try:
            # 1. 악플 메시지 즉시 삭제
            await message.delete()

            # 2. 경고 스택 1 증가
            user_id = message.author.id
            user_warnings[user_id] = user_warnings.get(user_id, 0) + 1
            current_warn = user_warnings[user_id]

            # 3-A. 5회 누적 시 -> 1분 타임아웃(채팅 금지) 처벌!
            if current_warn >= MAX_WARNINGS:
                user_warnings[user_id] = 0  # 경고 초기화
                
                # ⭐️ 디스코드 공식 1분 타임아웃 실행
                await message.author.timeout(timedelta(minutes=1), reason="악플/욕설 5회 누적")
                
                punish_msg = await message.channel.send(
                    f"⛔ **{message.author.mention}**님, 부적절한 표현이 **{MAX_WARNINGS}회 누적**되어 **1분간 채팅이 금지(타임아웃)**되었습니다!"
                )
                await punish_msg.delete(delay=8.0)
                print(f"🚫 [타임아웃 처벌] 유저: {message.author} (1분 채금)")

            # 3-B. 1~4회 경고 단계 -> 자폭 경고장 발송
            else:
                warning_msg = await message.channel.send(
                    f"🚨 **{message.author.mention}**님, 부적절한 표현이 감지되어 메시지가 삭제되었습니다!\n"
                    f"*(사유: **{label}**, 누적 경고: **{current_warn}/{MAX_WARNINGS}회**)*"
                )
                await warning_msg.delete(delay=5.0)
                print(f"[삭제 검거] 유저: {message.author} | 경고: {current_warn}/{MAX_WARNINGS} | 사유: {label}")

        except discord.errors.Forbidden:
            print("⚠️ 봇에게 '멤버 타임아웃/관리(Moderate Members)' 권한이 없습니다!")

BOT_TOKEN = CLEANBOT

if not BOT_TOKEN:
    print("❌ 에러: CLEANBOT 토큰을 찾을 수 없습니다!")
else:
    try:
        client.run(BOT_TOKEN)
    except KeyboardInterrupt:
        print("\n🛑 주피터 노트북에서 정지되었습니다. 클린봇 순찰을 안전하게 종료합니다!")
```

```python
import json
import random
import time
import re
import pandas as pd
from transformers import pipeline

# ====================================================================
# [1] 300개 균형 벤치마크 데이터셋 생성 (100개 정상 + 200개 욕설/우회)
# ====================================================================
clean_seeds = [
    "오늘 점심 메뉴로 돈까스 어때?", "이번 주말에 친구들이랑 축구하기로 했어.",
    "프로젝트 마감 기한이 얼마 안 남아서 야근해야겠다.", "새로 산 운동화 신고 출근했더니 기분 좋네.",
    "수박 먹을 때 수박 씨 골라내는 게 제일 귀찮아.", "포도 씨를 발라서 맛있게 먹자.",
    "참외 씨는 그냥 먹어도 배탈 안 난대.", "그 실패가 우리 팀의 새로운 시발점이 되었어.",
    "어제 본 영화 개재밌고 감동적이었음 꼭 봐라.", "과제 다 끝내고 나니까 진짜 살 것 같다.",
    "내일 아침 9시까지 3층 회의실로 모여주세요.", "지하철 연착돼서 지각할 뻔했네 휴.",
    "주말에 가족들이랑 제주도 여행 가기로 계획 짰어.", "조카 신발 선물해 줬더니 너무 좋아하더라.",
    "신발끈이 자꾸 풀려서 다시 꽉 묶었어.", "할머니께서 애미야 국이 짜다 하시더라.",
    "퇴근길에 마트 들러서 저녁거리 사가야겠네.", "새로 나온 신메뉴 음료수 맛이 꽤 괜찮네.",
    "오늘 날씨가 화창해서 산책하기 딱 좋은 날씨다.", "비가 많이 와서 우산 챙겨 나가야겠어.",
    "커피 한잔 마시면서 잠 좀 깨야겠다.", "다음 주 월요일에 정기 회의 있는 거 맞지?",
    "컴퓨터 모니터 새로 샀는데 화면 진짜 넓다.", "오늘 저녁은 집에서 삼겹살 구워 먹을래."
]

overt_seeds = [
    "야 이 개새끼야 닥쳐라", "진짜 병신 같은 소리 하고 있네", "너 씨발 진짜 맞고 싶냐?",
    "지랄하지 말고 꺼져", "존나 짜증나게 굴지 마라", "미친 새끼가 선 넘네",
    "조까는 소리 하지 말고 가만히 있어", "싸가지 없는 새끼", "대가리에 똥만 찼나",
    "니애미 니애비한테나 그렇게 말해라", "느금마 밥이나 차려드려라", "앰창 까고 맹세하냐?",
    "상종도 못할 쓰레기 새끼", "아가리 닥치고 가만히 있어라", "저 새끼 진짜 죽여버리고 싶네"
]

bypassed_seeds = [
    "야 이 tlqkf 놈아 저리 가", "wlfkf 떨지 말고 조용히 해", "qudtls 같은 짓 좀 그만해라", "tlqkfsus아 헛소리 마라",
    "아침부터 버스 놓치고 ㅅㅂ 망했네", "아군 정글러 플레이 보고 ㅂㅅ 소리 절로 나옴", "내 뒤에서 ㅈ1ㄹ 좀 하지 마",
    "일 처리 그따구로밖에 못하냐 18년아", "새벽에 술 취해서 전화 거는 18새끼 극혐이다", "선 넘지 마라 ㅆ1ㅂ 진짜",
    "채팅창에서 자꾸 지~랄 거리는 애 차단하자", "미_친_새_끼야 적당히 해라", "주차 이따위로 할래? 개@새@끼야",
    "선배들한테 욕먹고 꺼 져 소리 들음", "야 이 ㅅㅣ1발 지각하겠다", "술 취해서 씗빪 거리던 놈 누구냐",
    "자꾸 뒤에서 ㅈㄲ는 소리 하지 마라", "보기 싫으니까 채팅창에서 ㄲㅈ", "운전 꼬라지 보니까 ㅁㅣㅊㅣㄴ놈이네"
]

advanced_seeds = [
    "너 진짜 뇌가 순수하네", "뇌에 보톡스 맞아서 주름이 펴졌냐?", "뇌가 우동사리라 생각이 없지?",
    "오늘부터 널 보니 진화설은 폐기해야 할 것 같다", "똥이라도 먹을 줄 알면 정화조 대신이라도 쓸 텐데",
    "칭찬 고마운데 그 나이 먹고 그러고 싶냐 진짜? ㅋㅋ", "너 혹시 학창시절에 은따 당했냐? 말투 왜 이럼",
    "진짜 지능에 문제 있는 거 스스로 인증하네", "가정교육을 참 독특하게 받으신 것 같아요^^",
    "갈비뼈 순서 바꿔줄까?", "턱주가리 위치 돌려놔 줄까?", "아가리 찢어버리기 전에 조용히 해라",
    "온몸에 여드름 갈색 흉터 가득한 문둥이놈아", "수박씨 발라먹을 놈아", "포도씨 발라먹을 새끼야",
    "참외씨 발라먹을 자식아", "읾값겖 갋짊맑셁욞 젅낡 맜없굵 밄칞 섁끾들읾갊"
]

benchmark_300 = []

def generate_variants(seed_list, target_count, is_swear, category_name):
    count = 0
    tails = ["", " 진짜", " ㅋㅋ", " ㅡㅡ", " ㅠㅠ", "!", " 아오", " 참나", " 맞지?"]
    while count < target_count:
        base = random.choice(seed_list)
        tail = random.choice(tails)
        text = (base + tail).strip()
        if not any(d['text'] == text for d in benchmark_300):
            benchmark_300.append({
                "id": len(benchmark_300) + 1,
                "text": text,
                "is_swear": is_swear,
                "category": category_name
            })
            count += 1

random.seed(42)
generate_variants(clean_seeds, 100, False, "일상 대화(정상)")
generate_variants(overt_seeds, 50, True, "일반 욕설")
generate_variants(bypassed_seeds, 75, True, "우회 욕설(영타/자모/숫자)")
generate_variants(advanced_seeds, 75, True, "돌려까기/신체협박/외계어")

random.shuffle(benchmark_300)
for idx, item in enumerate(benchmark_300, 1):
    item["id"] = idx

# 📁 팀원/팀장님께 전달할 CSV 및 JSON 저장
df = pd.DataFrame(benchmark_300)
df.to_csv("benchmark_300.csv", index=False, encoding="utf-8-sig")
with open("benchmark_300.json", "w", encoding="utf-8") as f:
    json.dump(benchmark_300, f, ensure_ascii=False, indent=2)

print("✅ [파일 저장 완료]")
print("  - benchmark_300.csv  (엑셀/팀원 전달용)")
print("  - benchmark_300.json (JSON 형식)\n")


# ====================================================================
# [2] 내 모델 엔진 (우리가 완성한 하이브리드 파이프라인)
# ====================================================================
ENG_KEY_MAP = {
    'q': 'ㅂ', 'w': 'ㅈ', 'e': 'ㄷ', 'r': 'ㄱ', 't': 'ㅅ',
    'y': 'ㅛ', 'u': 'ㅕ', 'i': 'ㅑ', 'o': 'ㅐ', 'p': 'ㅔ',
    'a': 'ㅁ', 's': 'ㄴ', 'd': 'ㅇ', 'f': 'ㄹ', 'g': 'ㅎ',
    'h': 'ㅗ', 'j': 'ㅓ', 'k': 'ㅏ', 'l': 'ㅣ',
    'z': 'ㅋ', 'x': 'ㅌ', 'c': 'ㅊ', 'v': 'ㅍ', 'b': 'ㅠ',
    'n': 'ㅜ', 'm': 'ㅡ',
    'Q': 'ㅃ', 'W': 'ㅉ', 'E': 'ㄸ', 'R': 'ㄲ', 'T': 'ㅆ',
    'O': 'ㅒ', 'P': 'ㅖ'
}

CHOSUNG = ['ㄱ', 'ㄲ', 'ㄴ', 'ㄷ', 'ㄸ', 'ㄹ', 'ㅁ', 'ㅂ', 'ㅃ', 'ㅅ', 'ㅆ', 'ㅇ', 'ㅈ', 'ㅉ', 'ㅊ', 'ㅋ', 'ㅌ', 'ㅍ', 'ㅎ']
JUNGSUNG = ['ㅏ', 'ㅐ', 'ㅑ', 'ㅒ', 'ㅓ', 'ㅔ', 'ㅕ', 'ㅖ', 'ㅗ', 'ㅘ', 'ㅙ', 'ㅚ', 'ㅛ', 'ㅜ', 'ㅝ', 'ㅞ', 'ㅟ', 'ㅠ', 'ㅡ', 'ㅢ', 'ㅣ']
JONGSUNG = ['', 'ㄱ', 'ㄲ', 'ㄳ', 'ㄴ', 'ㄵ', 'ㄶ', 'ㄷ', 'ㄹ', 'ㄺ', 'ㄻ', 'ㄼ', 'ㄽ', 'ㄾ', 'ㄿ', 'ㅀ', 'ㅁ', 'ㅂ', 'ㅄ', 'ㅅ', 'ㅆ', 'ㅇ', 'ㅈ', 'ㅊ', 'ㅋ', 'ㅌ', 'ㅍ', 'ㅎ']

COMPLEX_BATCHIM_MAP = {
    'ㄳ': 'ㄱ', 'ㄵ': 'ㄴ', 'ㄶ': 'ㄴ', 'ㄺ': 'ㄱ', 
    'ㄻ': 'ㅁ', 'ㄼ': 'ㅂ', 'ㄽ': 'ㄹ', 'ㄾ': 'ㄹ', 
    'ㄿ': 'ㅂ', 'ㅀ': 'ㄹ', 'ㅄ': 'ㅂ'
}

SLANG_MAP = {
    "개색구": "개새끼", "개색기": "개새끼", "개새꾸": "개새끼",
    "시베리아 귤껍데기": "씨발", "븅신": "병신", "븡신": "병신", "조까": "좆까", "밑칲": "미친",
    "ㅅㅂ": "시발", "ㅆㅂ": "씨발", "ㅂㅅ": "병신", "ㅈㄹ": "지랄",
    "ㅈㄲ": "좆까", "ㄲㅈ": "꺼져", "ㅁㅊ": "미친", "ㅈㄴ": "존나",
    "18년": "십팔년", "18새끼": "십팔새끼",
    "섁끾": "새끼", "색끾": "새끼", "섁끼": "새끼",
    "밄칞": "미친", "밄친": "미친", "밀칞": "미친",
    "젅낡": "존나", "전낡": "존나"
}

HARD_SWEAR_WORDS = ["느금마", "느검마", "니애미", "니애비", "너네애미", "앰창"]

SAFE_PHRASES = [
    r'(수박|포도|참외|호박|사과|감|복숭아|자두|석류|과일|고추)\s*씨',
    r'조카\s*신발',
    r'시발점',
    r'신발끈',
    r'애미야',
    r'어미야'
]
PEJORATIVE_TARGETS = r'(놈|년|새끼|새키|새퀴|자식|인간|련|넘)'

SARCASTIC_CONTEXT_PATTERNS = [
    (r'(밄|밀)칞?.*(섁|끾|새끼)|섁끾|섁끼', "외계어 우회 욕설", 85.0),
    (r'(그\s*나이\s*(먹고|처먹고)|나잇값|나이값).*(그러고|왜\s*그|부끄|한심|참)', "연령/인격 비하", 55.0),
    (r'(학창시절|학창\s*시절).*(은따|왕따|찐따|따돌림)|(은따|왕따|찐따)\s*(당했|출신|냐|같)', "따돌림/인격 비하", 50.0),
    (r'뇌.*(순수|보톡스|우동|장식|정지|텅)', "지능 돌려까기", 55.0),
    (r'(진화설|정화조|재활용).*쓸텐데|폐기', "인격 모독", 50.0),
    (r'(갈비뼈|뼈|턱|이빨|아가리).*(순서|바꿔|돌려|뽑)', "폭력/신체협박", 75.0),
    (r'(문둥이|한센|문둥병)', "질병/신체 비하", 60.0)
]

def check_is_really_safe(text):
    for phrase in SAFE_PHRASES:
        if re.search(phrase, text):
            if re.search(rf"{phrase}.*{PEJORATIVE_TARGETS}", text):
                return False
            return True
    return False

def convert_eng_to_jamo(text):
    def replace_token(match):
        eng = match.group()
        return "".join([ENG_KEY_MAP.get(ch, ch) for ch in eng])
    return re.sub(r'[a-zA-Z]+', replace_token, text)

def compose_syllable(cho, jung, jong=''):
    c_idx = CHOSUNG.index(cho)
    j_idx = JUNGSUNG.index(jung)
    t_idx = JONGSUNG.index(jong) if jong in JONGSUNG else 0
    return chr(0xAC00 + (c_idx * 21 + j_idx) * 28 + t_idx)

def combine_jamo(text):
    chars = list(text)
    result = []
    i = 0
    n = len(chars)
    while i < n:
        c = chars[i]
        if c in CHOSUNG and i + 1 < n and chars[i+1] in JUNGSUNG:
            cho = c
            jung = chars[i+1]
            jong = ""
            if i + 2 < n and chars[i+2] in JONGSUNG and chars[i+2] != "":
                if i + 3 < n and chars[i+3] in JUNGSUNG:
                    jong = ""
                    i += 2
                else:
                    jong = chars[i+2]
                    i += 3
            else:
                i += 2
            result.append(compose_syllable(cho, jung, jong))
        else:
            result.append(c)
            i += 1
    return "".join(result)

def clean_bypassed_text(text):
    text = convert_eng_to_jamo(text)
    text = re.sub(r'18([년놈새끼자식])', r'십팔\1', text)
    for slang, clean in SLANG_MAP.items():
        text = text.replace(slang, clean)
    text = re.sub(r'[\d~!@#$%^&*()_+|<>?:{}\[\]\-_.]', '', text)
    text = combine_jamo(text)
    
    normalized_chars = []
    for char in text:
        if '가' <= char <= '힣':
            code = ord(char) - 0xAC00
            cho = CHOSUNG[code // (21 * 28)]
            jung = JUNGSUNG[(code % (21 * 28)) // 28]
            jong = JONGSUNG[code % 28]
            if jong in COMPLEX_BATCHIM_MAP:
                jong = COMPLEX_BATCHIM_MAP[jong]
            normalized_chars.append(compose_syllable(cho, jung, jong))
        else:
            normalized_chars.append(char)
    text = "".join(normalized_chars)
    text = text.replace("씨빨", "씨발").replace("시빨", "시발")
    return text

print("⏳ AI 모델 로딩 중...")
classifier = pipeline("text-classification", model="smilegate-ai/kor_unsmile", top_k=None)
_ = classifier("테스트")
print("✅ 클린봇 엔진 준비 완료!\n")

def predict_my_model(text):
    is_hard_swear = any(bad in text for bad in HARD_SWEAR_WORDS)
    is_safe = check_is_really_safe(text)
    restored = clean_bypassed_text(text)
    
    results = classifier(restored)[0]
    toxic_labels = [r for r in results if r['label'] != 'clean']
    top_toxic = max(toxic_labels, key=lambda x: x['score'])
    label, score = top_toxic['label'], top_toxic['score'] * 100
    
    for pattern, boost_label, boost_score in SARCASTIC_CONTEXT_PATTERNS:
        if re.search(pattern, restored):
            score = min(score + boost_score, 99.9)
            label = boost_label
            break

    if is_hard_swear or any(bad in restored for bad in HARD_SWEAR_WORDS):
        return True, "패드립/가족", 99.99
    elif is_safe and score < 90:
        return False, "정상", score
    elif score >= 50:
        return True, label, score
    else:
        return False, "정상", score


# ====================================================================
# [3] 300개 대규모 벤치마크 평가 실행
# ====================================================================
print("⏳ 내 모델로 300개 문장 정밀 평가를 시작합니다...\n")

categories = ["일상 대화(정상)", "일반 욕설", "우회 욕설(영타/자모/숫자)", "돌려까기/신체협박/외계어"]
stats = {cat: {"correct": 0, "total": 0} for cat in categories}

total_time = 0.0
total_correct = 0

for data in benchmark_300:
    text = data["text"]
    actual = data["is_swear"]
    cat = data["category"]
    
    t0 = time.time()
    pred, label, score = predict_my_model(text)
    elapsed = time.time() - t0
    total_time += elapsed
    
    is_correct = (pred == actual)
    if is_correct:
        stats[cat]["correct"] += 1
        total_correct += 1
    stats[cat]["total"] += 1

# ====================================================================
# 📊 [최종 성적표 출력]
# ====================================================================
print("=" * 80)
print("📊 [내 모델: 300개 종합 벤치마크 최종 성적표]")
print("=" * 80)
print(f"{'카테고리':<28} | {'데이터 수':<8} | {'정답 수':<8} | {'정확도'}")
print("-" * 80)

for cat in categories:
    cnt = stats[cat]["total"]
    corr = stats[cat]["correct"]
    acc = (corr / cnt) * 100
    print(f"{cat:<28} | {cnt:<8} | {corr:<8} | {acc:>6.1f}%")

avg_latency = (total_time / len(benchmark_300)) * 1000
total_acc = (total_correct / len(benchmark_300)) * 100

print("=" * 80)
print(f"🏆 전체 종합 정확도 (300개 중) : {total_correct} / {len(benchmark_300)} ({total_acc:.2f}%)")
print(f"⚡ 평균 처리 속도 (Latency)   : 건당 {avg_latency:.2f}ms (약 {avg_latency/1000:.3f}초)")
print("=" * 80)
```
