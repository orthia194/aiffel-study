```python
import torch
import torch.nn as nn

batch_size = 64
pic = torch.zeros(batch_size, 3, 1080, 1920)    # PyTorch는 (Batch, Channel, Height, Width)
                                                # 순서의 channels-first를 사용합니다.

print("입력 이미지 데이터:", pic.shape)
conv_layer = nn.Conv2d(in_channels=3,
                       out_channels=16,
                       kernel_size=(5, 5),
                       stride=5,
                       bias=False)
conv_out = conv_layer(pic)

print("\nConvolution 결과:", conv_out.shape)
print("Convolution Layer의 Parameter 수:", sum(p.numel() for p in conv_layer.parameters()))

flatten_out = nn.Flatten()(conv_out)
print("\n1차원으로 펼친 데이터:", flatten_out.shape)

linear_layer = nn.Linear(in_features=flatten_out.shape[1], out_features=1, bias=False)
linear_out = linear_layer(flatten_out)

print("\nLinear 결과:", linear_out.shape)
print("Linear Layer의 Parameter 수:", sum(p.numel() for p in linear_layer.parameters()))
```
